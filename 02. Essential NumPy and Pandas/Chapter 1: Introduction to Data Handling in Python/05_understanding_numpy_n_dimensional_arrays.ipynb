# **Understanding NumPy N-dimensional Arrays**

- Cấu trúc dữ liệu cốt lõi của NumPy là **mảng N chiều (N-dimensional array)**, thường được gọi là `ndarray`.

- Có thể xem `ndarray` như một **vùng chứa dữ liệu (data container)** có hiệu suất cao và tính linh hoạt lớn, được thiết kế chuyên biệt cho **dữ liệu số (numerical data)**

- Mặc dù Python lists có khả năng lưu trữ nhiều loại dữ liệu và phù hợp với nhiều mục đích khác nhau, chúng không được tối ưu hóa cho các **phép toán số quy mô lớn (large-scale numerical operations)** thường gặp trong DS và ML.

- NumPy arrays được phát triển nhằm khắc phục những hạn chế này.

## **1. What makes an `ndarray` special?**



### **1.1. Homogeneous Data Type**:

- Không giống với Python lists, vốn có thể chứa các phần tử thuộc nhiều **data types** khác nhau trong cùng một list, tất cả phần tử bên trong một NumPy array thường phải có cùng một **data type**

- VD: Một python list có thể chứa đồng thời: `python_list = [10, "Alice", 3.14, True]`: List có `Số nguyên`, `Chuỗi ký tự`, `Số thực`, `boolean value`.

- Ngược lại, trong một NumPy array, các phần tử thường được chuyển đổi về cùng một data type phù hợp

- Tính chất này được gọi là **tính đồng nhất kiểu dữ liệu (data type homogeneity)** và là yếu tố quan trọng tạo nên hiệu suất của NumPy.

- Khi NumPy biết rằng mọi phần tử trong mảng đều là cùng một loại dữ liệu, chẳng hạn như **số thực dấu phẩy động 64 bit (64-bit floating-point number)**, thư viện có thể sử dụng **mã C cấp thấp đã được tối ưu hóa (optimized low-level C code)** để thực hiện các phép tính.  

- Cách xử lý này giúp giảm phần lớn **chi phí xử lý bổ sung (overhead)** liên quan đến **kiểu dữ liệu động (dynamic typing)** của Python.

### **1.2. Fixed Size**

- Khi một NumPy array được tạo, kích thước của nó thường được xem là cố định.

- NumPy vẫn cung cấp các hàm để thay đổi kích thước hoặc cấu trúc của mảng. Tuy nhiên, trong nhiều trường hợp, các thao tác này không trực tiếp thay đổi mảng ban đầu.

- Thay vào đó, NumPy thường thực hiện các bước sau:

  - Tạo một mảng mới.
  - Sao chép dữ liệu từ mảng cũ.
  - Bổ sung hoặc loại bỏ phần tử theo yêu cầu.
  - Trả về mảng mới.

- Điều này khác với việc thay đổi kích thước tại chỗ (in-place resizing) trên chính vùng nhớ ban đầu.

- Đặc tính kích thước tương đối cố định giúp NumPy tối ưu hóa quá trình:

  - **Cấp phát bộ nhớ (memory allocation)**
  - **Truy cập bộ nhớ (memory access)**
  - Thực hiện phép toán trên mảng

### **1.3. Multidimensionality**

- Đúng như tên gọi, `ndarray` có thể chứa nhiều chiều dữ liệu.
  -  **Mảng một chiều (one-dimensional array)** tương tự như một danh sách hoặc một vector. VD: `[10, 20, 30, 40, 50]` -> **one-dimensional array**
  - **Mảng hai chiều (two-dimensional array)** tương tự như một bảng hoặc một ma trận gồm hàng và cột. Ví dụ:
  ```python
  [[1, 2, 3],
    [4, 5, 6]]
  ```
    - Mảng trên có:
      - 2 hàng
      - 3 cột

  - **Mảng n chiều (n-dimensional array)**:
    - NumPy arrrays có thể có ba, bốn hoặc nhiều chiều hơn. Khả năng này cho phép NumPy biểu diễn các tập dữ liệu phức tạp, chẳng hạn như ảnh màu.

    - Một ảnh màu có thể được biểu diễn bằng mảng ba chiều:
      - Chiều thứ nhất: chiều cao (height)
      - Chiều thứ hai: chiều rộng (width)
      - Chiều thứ ba: các kênh màu (color channels)
    - Ví dụ, một ảnh RGB có kích thước `1080 x 1920` có thể được biểu diễn bằng một mảng có hình dạng: `(1080, 1920, 3)`. Trong đó, giá trị `3` là đại diện cho ba kênh màu: `Đỏ`, `Xanh lá`, `Xanh dương`.

    - Mỗi chiều trong một NumPy array được gọi là một **trục (axis)**.VD:
      - Mảng một chiều có một trục.
      - Mảng hai chiều có hai trục.
      - Mảng ba chiều có ba trục.

### **1.4. Efficiency**

- NumPy array thường lưu trữ dữ liệu trong một **khối bộ nhớ liên tục (contiguous block of memory)**.

- Ngoài ra, các phép toán trên mảng được thực hiện bằng **mã C đã biên dịch (compiled C code)** thay vì xử lý từng phần tử thông qua vòng lặp Python. Nhờ đó, các phép toán trên `ndarray` thường nhanh hơn đáng kể so với những phép toán tương đương được thực hiện trên danh sách Python. Khả năng này thường được gọi là **vector hóa (vectorization)**

- **Vectorization** cho phép thực hiện một phép toán trên toàn bộ tập dữ liệu mà không cần viết các **vòng lặp tường minh (explicit loops)**. Ví dụ, để nhân tất cả phần tử trong một mảng NumPy với `2`, có thể viết: `numpy_array * 2` thay vì sử dụng vòng lặp: `for value in python_list`
  - Cách tiếp cận vector hóa giúp mã nguồn:

    - Ngắn gọn hơn
    - Dễ đọc hơn
    - Nhanh hơn
    - Phù hợp hơn với dữ liệu quy mô lớn

- Đoạn mã sau minh họa cách chuyển một danh sách Python thành một đối tượng `ndarray`


````python
# Nhập thư viện NumPy với bí danh phổ biến là 'np'
import numpy as np

# Tạo một danh sách Python đơn giản
python_list = [10, 20, 30, 40, 50]

# Chuyển danh sách thành một NumPy ndarray
numpy_array = np.array(python_list)

# In mảng và kiểu của đối tượng
print("NumPy Array:", numpy_array)
print("Type:", type(numpy_array))
````

- Kết quả: `[10 20 30 40 50]` trông tương tự một danh sách Python nhưng không có dấu phẩy giữa các phần tử. - Đây là **biểu diễn chuỗi tiêu chuẩn (standard string representation)** của một mảng NumPy một chiều.

- Dòng kết quả: `<class 'numpy.ndarray'>` xác nhận rằng biến `numpy_array` là một đối tượng thuộc lớp `numpy.ndarray`. - Nói cách khác, dữ liệu ban đầu đã được chuyển từ một danh sách Python thông thường thành cấu trúc mảng chuyên biệt của NumPy.

- Những đặc điểm quan trọng của `ndarray` bao gồm:
