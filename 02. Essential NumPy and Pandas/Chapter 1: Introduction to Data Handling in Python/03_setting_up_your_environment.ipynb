# **Setting Up Your Environment**

- Trước khi bắt đầu làm việc với NumPy và Pandas, bạn cần cài đặt hai thư viện này trên máy tính. NumPy và Pandas không được tích hợp sẵn trong Python theo mặc định, vì vậy người dùng cần chủ động cài đặt thêm.
- Quá trình này được thực hiện thông qua các công cụ gọi là **trình quản lý gói (package managers)**. Những công cụ này hỗ trợ tải xuống, cài đặt và quản lý các **thư viện phần mềm (software libraries)** cần thiết cho dự án.

- Có hai phương pháp phổ biến để thiết lập môi trường làm việc (working environment) cho các dự án Khoa học dữ liệu bằng Python:
  1. Sử dụng **bản phân phối Anaconda (Anaconda distribution)**
  2. Sử dụng **trình cài đặt gói tiêu chuẩn của Python (Python standard package installer)** là `pip`

- Đối với người mới bắt đầu, Anaconda thường được khuyến nghị vì công cụ này giúp đơn giản hóa đáng kể quá trình cài đặt và quản lý môi trường.

## **1. Option 1: Installing Anaconda (Recommended)**

- Anaconda là một **bản phân phối mã nguồn mở và miễn phí (free and open-source distribution)** của Python và R, được thiết kế chuyên biệt cho t**ính toán khoa học (scientific computing)** và Khoa học dữ liệu (Data Science).

- Anaconda được tích hợp sẵn:
  - Python
  - Nhiều thư viện quan trọng, bao gồm NumPy và Pandas
  - Jupyter Notebook
  - Trình quản lý gói riêng có tên là `conda`




### **1.1. Why Anaconda?**

- **Convenience**:
  - Anaconda cho phép cài đặt Python cùng hàng trăm **gói Khoa học dữ liệu phổ biến (popular data science packages)** chỉ trong một lần.

  - NumPy và Pandas thường đã được tích hợp sẵn trong Anaconda. Vì vậy, người dùng có thể không cần cài đặt riêng từng thư viện.

- **Environment Management**:
  - `conda` giúp người dùng dễ dàng tạo các **môi trường độc lập (isolated environments)** cho từng dự án.
  - Việc sử dụng môi trường độc lập giúp ngăn chặn **xung đột phiên bản thư viện (library version cònlicts)** giữa các dự án khác nhau.

  - VD: Một dự án có thể yêu cần Pandas phiên bản cũ, trong khi một dự án khác cần phiên bản mới. Việc tách riêng môi trường giúp cả hai dự án có thể hoạt dodnogoj ổn định trên cùng một mánh tính.

- **Cross-Platform**: Anaconda có thể hoạt động trên nhiều hệ điều hành (operating systems), bao gồm:
  - Windows
  - macOS
  - Linux

## **1.2. Anaconda installation steps**

1. **Download**:
    - Truy cập trang web chính thức của **Anaconda Distribution** và tải xuống bộ cài đặt phù hợp với hệ điều hành đang sử dụng (Địa chỉ:https://www.anaconda.com/products/distribution). Chọn phiên bản **Python 3.x**. Nhìn chung, nên sử dụng **phiên bản ổn định mới nhất (latest stable version)**.

2. **Install**:
    - Khởi chạy tệp cài đặt vừa tải xuống và thực hiện theo các hướng dẫn xuất hiện trên màn hình.
    - Nên giữ nguyên các **thiết lập mặc định (default settings)**, trừ khi bạn có yêu cầu cụ thể đối với môi trường phát triển.
    - Một lựa chọn quan trọng trong quá trình cài đặt, đặc biệt trên Windows, là có thêm Anaconda vào **biến môi trường PATH của hệ thống (system PATH environment variable)** hay không.
    - Thông thường, trình cài đặt khuyến nghị không thêm Anaconda trực tiếp vào `PATH`. Thay vào đó, người dùng nên sử dụng:
      - **Anaconda Prompt** trên Windows
      - Terminal thông thường trên macOS hoặc Linux
    - Đây là lựa chọn an toàn vì nó hạn chế các xung đột với những phiên bản Python khác đã được cài đặt trên hệ thống.

3. **Verify (Optional)**:
    - Sau khi cài đặt hoàn tất, mở Anaconda Prompt trên Windows hoặc Terminal trên macOS và Linux.
    - Nhập lệnh sau để kiểm tra xem conda đã được cài đặt thành công hay chưa: `conda --version`.
    - Nếu quá trình cài đặt thành công, hệ thống sẽ hiển thị **số phiên bản conda (conda version number)**.

4. **Check/Install NumPy and Pandas**:
    - Anaconda thường đã bao gồm NumPy và Pandas. Bạn có thể kiểm tra bằng lệnh: `conda list numpy pandas`.

    - Nếu NumPy và Pandas xuất hiện trong danh sách, môi trường đã sẵn sàng sử dụng.
    - Trong trường hợp hợp các thư viện chưa được cài đặt, hãy chạy:
    
      ```py
      # To install if missing:
      conda install numpy pandas jupyterlab

      # To update existing packages:
      conda update numpy pandas jupyterlab
      ```

    - Trong các lệnh trên, `jupyterlab` cũng được cài đặt vì đây là **môi trường Jupyter Notebook (Jupyter Notebook environment)** sẽ được sử dụng trong các phần tiếp theo.


## **2. Option 2: Using pip and Virtual Environments**

- Nếu Python đã được cài đặt trên hệ thống, chẳng hạn như được tải từ `python.org`, và bạn không muốn sử dụng Anaconda, bạn có thể sử dụng `pip`.

- `pip` là **trình cài đặt gói mặc định của Python (default Python package installer)**.

### **2.1. Why pip?**

- **Standard**:
  - `pip` là công cụ quản lý và cài đặt gói tiêu chuẩn trong hệ sinh thái Python.
  - Phần lớn các thư viện Python có thể được cài đặt trực tiếp từ **Python Package Index (PyPI)** bằng `pip`

- **Minimal**:
  - `pip` cho phép tạo một môi trường phát triển nhẹ hơn nếu bạn không cần toàn bộ các công cụ và thư viện được tích hợp trong Anaconda.

  - Thay vì cài đặt hàng trăm thư viện, người dùng chỉ cần cài đặt những gói thực sự cần thiết cho dự án.

### **2.2. Best Practice: Virtual Environments**

- Khi sử dụng `pip`, bạn nên kết hợp với **môi trường ảo (virtual environment)**.

- Virtual environment là một thư mục độc lập, chứa:
  - Một phiên bản Python cụ thể
  - Một tập hợp thư viện riêng
  - Các tệp cấu hình cần thiết cho môi trường đó
- Mục đích chính của virtual environment là ngăn chặn xung đột giữa các thư viện được sử dụng trong những dự án khác nhau.

- Python hỗ trợ sẵn việc tạo virtual environment thông qua mô-đun `venv`.

#### **Steps**:

- **Ensure Python and pip**:
  - Trước tiên, hãy đảm bảo rằng Python 3 và pip đã được cài đặt trên hệ thống.

  - Mở Terminal hoặc dấu **nhắc lệnh (command prompt)** và chạy:

    ```py
      python --version
      # or maybe python3 --version
      pip --version
      # or maybe pip3 --version
    ```
  - Nếu các lệnh trên hoạt động và hiển thị số phiên bản tương ứng, Python và `pip` đã sẵn sàng.

  - Nếu không, bạn cần cài đặt Python từ trang web chính thức: https://www.python.org. `pip` thường được tích hợp sẵn trong Python từ phiên bản 3.4 trở lên.

- **Create a Virtual Environment**:
  - Trong Terminal, di chuyển đến **thư mục dự án (project directory)** hoặc tạo một thư mục mới. Sau đó chạy lệnh:
    ```py
    # Replace 'myenv' with your preferred environment name
    python -m venv myenv
    ```
  - Trong đó, `myenv` là tên của môi trường ảo. Bạn có thể thay thế bằng một tên khác phù hợp với dự án. VD: `python -m venv data-science-env`. Lệnh này tạo một thư mục có tên `myenv`, chứa các tệp cần thiết cho môi trường Python độc lập.

- **Activate the Environment**:
  - Trước khi cài đặt các thư viện, bạn cần **kích hoạt môi trường (activate the environment)**.
    - **On macOS and Linux**: `source myenv/bin/activate`
    - **On Windows (Command Prompt)**: `myenv\Scripts\activate.bat`
    - **On Windows (PowerShell)**: `myenv\Scripts\Activate.ps1`

  - Nếu **PowerShell** không cho phép chạy tệp kích hoạt, bạn có thể cần điều **chỉnh chính sách thực thi (execution policy)** bằng lệnh: `Set-ExecutionPolicy -ExecutionPolicy RemoteSigned -Scope Process`. Sau đó, chạy lại: `myenv\Scripts\Activate.ps1`

  - Khi môi trường được kích hoạt thành công, dấu nhắc **Terminal (terminal prompt)** thường thay đổi và hiển thị tên môi trường ở đầu dòng. `(myenv) C:\my-project>`. Điều này cho biết môi trường `myenv` đang hoạt động.

- **Install Libraries**:
  - Sau khi kích hoạt môi trường ảo, sử dụng pip để cài đặt NumPy, Pandas và JupyterLab: `pip install numpy pandas jupyterlab`.
  - `pip` sẽ tải xuống và cài đặt các thư viện vào môi trường ảo đang hoạt động. Các thư viện này sẽ không ảnh hưởng đến môi trường Python toàn hệ thống hoặc các dự án khác.

- **Deactivate (When Done)**: Sau khi hoàn thành công việc, bạn có thể **hủy kích hoạt môi trường (deactivate the environment)** bằng lệnh: `deactivate`. Sau khi chạy lệnh này, Terminal sẽ quay trở lại môi trường Python mặc định của hệ thống.

## **3. Verifying Your Installation**

- Bất kể bạn sử dụng Anaconda hay pip, việc kiểm tra xem các thư viện đã được cài đặt chính xác hay chưa là một thực hành tốt.

- **Start a Python Interpreter**: Mở Terminal hoặc Anaconda Prompt. Nếu sử dụng môi trường ảo, hãy đảm bảo môi trường đã được kích hoạt. Nhập: `python`. Sau đó nhấn Enter. Nếu Python khởi động thành công, bạn sẽ nhìn thấy **dấu nhắc Python (Python prompt)** (`>>>`)

- **Import Libraries**: Nhập lần lượt các lệnh sau:
  ```python
  import numpy as np
  import pandas as pd

  print(f"NumPy version: {np.__version__}")
  print(f"Pandas version: {pd.__version__}")
  ```
  - Trong đó:
    - `np` là bí danh (alias) thường được sử dụng cho NumPy.
    - `pd` là bí danh thường được sử dụng cho Pandas.

- **Check Output**: Nếu các lệnh được thực thi mà không xuất hiện lỗi và hệ thống hiển thị số phiên bản của NumPy và Pandas, quá trình cài đặt đã thành công.

- **Exit Interpreter**: Để thoát khỏi **trình thông dịch Python (Python interpreter)**, nhập: `exit()`. Sau đó nhấn Enter.
