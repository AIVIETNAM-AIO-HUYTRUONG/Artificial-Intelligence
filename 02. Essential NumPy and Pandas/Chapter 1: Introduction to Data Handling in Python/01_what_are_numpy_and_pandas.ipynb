# **What are Numpy and Pandas?**

- NumPy và Pandas là hai **thư viện nền tảng (foundational libraries)** trong **hệ sinh thái Python (Python ecosystem)** và thực hiện các tác vụ **phân tích dữ liệu (data analysis)**.

- có thể xem NumPy và Pandas như những bộ công cụ chuyên dụng để làm việc với tập hợp các con số, **bảng dữ liệu (data tables)** hoặc **dữ liệu chuỗi thời gian (time series data)**. Đây là những dạng dữ liệu thường xuyên xuất hiện trong **khoa học dữ liệu (data science)** và các ứng dụng **trí tuệ nhân tạo (artificial intelligence)**.

## **1. NumPy: The Foundation for Numerical Computing**

- NumPy, viết tắt **Numerical Python**, là lớp nền tảng cho phần lớn **hệ thống tính toán khoa học (scientifc computing stack)** trong Python.

- Thành phần quan trong nhất mà NumPy cung cấp là đối tượng `ndarray`, còn được gọi là **mảng N chiều (N-dimensional array)**. Điều gì làm cho array trong NumPy trở nên đặc biệt:

  - **Hiệu suất (Efficiency)**:
    - Mảng NumPy được triển khai bằng ngôn ngữ C và được tối ưu hóa cho **các phép toán số học (numerical operations)**. Các phép tính được thực hiện trên từng phần tử của NumPy arrays thường nhanh hơn đáng kể so với các phép tính tương đương trên **Python list** thông thường.

    - Tốc độ xử lý này đặc biệt quan trọng khi làm việc với các **tập dữ liệu lớn (large datasets)**

  -  **Hiệu quả sử dụng bộ nhớ (Memory Usage)**: NumPy arrays có khả năng sử dụng bộ nhớ hiệu quả hơn Python list, đặc biệt khi xử lý một **large datasets**. Nguyên nhân là vì các phần tử trong NumPy array thường có cùng một **kiểu dữ liệu (data type)** và được lưu trữ liên tiếp trong bộ nhớ. Cách tổ chức này được gọi là **lưu trữ liên tục trong bộ nhớ (contiguous memory storage)**.

  - **Chức năng toán học phong phú (Functionality)**: NumPy cung cấp một tập hợp toàn diện các **hàm toán học cấp cao (high-level mathematical functions)** có thể hoạt động trực tiếp trên mảng.

    - Các chức năng này bao gồm:
      - Các phép toán **đại số tuyến tính (linear algebra)**
      - **Biến đổi Fourier (Fourier transform)**
      - **Sinh số ngẫu nhiên (random number generation)**
      - Các công cụ phục vụ **phân tích thống kê (statistical analysis)**

- Về cơ bản, khi cần thực hiện các phép toán trên những khối dữ liệu số như **vector (vector)**, **ma trận (matrix)** hoặc **tensor nhiều chiều (higher-dimensional tensor)**, NumPy cung cấp các đối tượng và hàm nền tảng để xử lý chúng một cách hiệu quả.

- NumPy cũng là nền móng để xây dựng nhiều thư viện data analysis và machine learning khác, trong đó có Pandas.

## **2. Pandas: Data Analysis and Manipulation Made Easier**

- Trong khi NumPy cung cấp nền **tảng tính toán số cấp thấp (low-level numerical computing)**, Pandas cung cấp các **cấu trúc dữ liệu cấp cao (high-level data structures)** và công cụ phân tích được thiết kế để tăng tính thực tiễn và sự thuận tiện khi sử dụng.

- Pandas đặc biệt phù hợp với **dữ liệu dạng bảng (tabular data)**, chẳng hạn như dữ liệu trong **bảng tính (spreadsheet)** hoặc bảng trong **cơ sở dữ liệu SQL (SQL database)**.

- Hai cấu trúc dữ liệu chính trong Pandas là `Series` và `DataFrame`:


### **2.1. Series**

- `Series`: là một **mảng một chiều có nhãn (one-dimensional labeled array)**. Cấu trúc này tương tự như một cột trong bản tính hoặc một vector dữ liệu đi kèm với một **index**
- Dưới đây là ví dụ minh họa cách khởi tạo và truy xuất `pandas.Series`:**văn bản in đậm**

```python
import pandas as pd
import numpy as np

# Khởi tạo Series từ một list
data = [10, 20, 30, 40]
index = ['a', 'b', 'c', 'd']
s = pd.Series(data, index=index, name="example_series")

print("Series object:")
print(s)

# Truy xuất dữ liệu qua nhãn (label-based)
print(f"\nGiá trị tại index 'b': {s['b']}")

# Series hỗ trợ các thao tác vector hóa của NumPy
print("\nSeries nhân đôi giá trị:")
print(s * 2)

# Kiểm tra kiểu dữ liệu (dtype)
print(f"\nKiểu dữ liệu của các phần tử: {s.dtype}")
```

### **2.2. DataFrame**

- `DataFrame` là một **cấu trúc dữ liệu hai chiều có nhãn (two-dimensional labeled data structure)**, bao gồm các hàng và các cột. Mỗi cột trong một `DataFrame` có thể chứa một kiểu dữ liệu khác nhau.

- Có thể hình dung DataFrame tương tự như:

  - Một bảng tính
  - Một bảng trong cơ sở dữ liệu SQL
  - Một **từ điển các đối tượng Series (dictionary of Series objects)**

- `DataFrame` là đối tượng được sử dụng phổ biến nhất trong Pandas.

### **2.3. Pandas excels**

- **Data Handling**:
  - Pandas cho phép người dùng dễ dàng đọc dữ liệu từ nhiều định dạng tệp (file formats) và nguồn dữ liệu khác nhau, chẳng hạn như:

    - Tệp **CSV (Comma-Separated Values)**
    - Tệp **Excel (Excel file)**
    - Tệp **JSON (JavaScript Object Notation)**
    - **Cơ sở dữ liệu SQL (SQL database)**

  - Sau khi xử lý, dữ liệu cũng có thể được ghi trở lại các định dạng tương ứng.

- **Data Cleaning**:
  - Pandas cung cấp nhiều công cụ để thực hiện **làm sạch dữ liệu (data cleaning)** và xử lý **dữ liệu thiếu (missing data)**.

  - Các thao tác phổ biến bao gồm:
    - Phát hiện các **giá trị thiếu NaN (Not a Number values)**.
    - Điền giá trị thay thế cho dữ liệu thiếu
    - Loại bỏ các hàng hoặc cột chứa dữ liệu thiếu
    - **Lọc hàng (row filtering)** theo điều kiện
    - **Biến đổi dữ liệu (data transformation)** sang định dạng phù hợp

- **Data Exploration and Analysis**:
  - Pandas cung cấp các phương thức để lựa chọn, phân chia và phân tích dữ liệu.
  
  - Các thao tác phổ biến bao gồm:
    - Lựa chọn một **tập con dữ liệu (data subset)**
    - **Cắt dữ liệu (data slicing)** theo hàng hoặc cột
    - **Nhóm dữ liệu (data grouping)** dựa trên một hoặc nhiều tiêu chí
    - Thực hiện các **phép tính (calculations)** và **phép tổng hợp (aggregation)**
    - Tính tổng, giá trị trung bình và số lượng
    - **Hợp nhất dữ liệu (data merging)**
    - **Kết nối dữ liệu (data joining)** từ nhiều tập dữ liệu khác nhau

- **Time Series**:
  - Pandas cung cấp các công cụ chuyên biệt để làm việc với **dữ liệu có dấu thời gian (time-stamped data)**, chẳng hạn như dữ liệu được ghi nhận theo ngày, tháng, năm, giờ hoặc phút.

  - Các công cụ này hỗ trợ hiệu quả cho nhiều bài toán như:

    - Phân tích dữ liệu tài chính
    - Theo dõi doanh số theo thời gian
    - Phân tích lưu lượng truy cập
    - **Dự báo xu hướng (trend forecasting)**

- NumPy cung cấp các **đối tượng mảng được tối ưu hóa (optimized array objects)** cùng với hệ thống **công cụ toán học (mathematical machinery)** cần thiết cho quá trình tính toán số.

- Pandas được xây dựng trên nền tảng này nhằm cung cấp các **cấu trúc dữ liệu linh hoạt và trực quan (flexible and intuitive data structures)**, bao gồm `Series` và `DataFrame`.

- Bên cạnh đó, Pandas còn cung cấp một tập hợp phong phú các hàm hỗ trợ:

  - **Nạp dữ liệu (data loading)**
  - **Làm sạch dữ liệu (data cleaning)**
  - **Biến đổi dữ liệu (data transformation)**
  - **Hợp nhất dữ liệu (data merging)**
  - **Phân tích dữ liệu (data analysis**

- Trong thực tế, NumPy và Pandas gần như luôn được sử dụng kết hợp với nhau khi làm việc với các **dự án lấy dữ liệu làm trọng tâm (data-centric projects)** trong Python.
