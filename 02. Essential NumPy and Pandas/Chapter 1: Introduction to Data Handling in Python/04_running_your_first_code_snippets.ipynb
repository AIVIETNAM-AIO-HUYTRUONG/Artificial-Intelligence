## **2. Your First NumPy Array**

- Đối tượng chính của NumPy là **mảng N chiều (N-dimensional array)**, thường được gọi là `ndarray`.

- Trước tiên, chúng ta sẽ tạo ra một **mảng một chiều (one-dimensional array)** đơn giản từ danh sánh Python thông thường.

````python
# Tạo một danh sách Python
my_list = [1, 2, 3, 4, 5]

# Tạo mảng NumPy từ danh sách
np_array = np.array(my_list)

# In mảng
print("My first NumPy array:")
print(np_array)

# In kiểu của đối tượng mảng
print("Type of the object:")
print(type(np_array))
````

- Trong mảng NumPy, các phần tử thường đoực hiển thị mà không có dấu phẩy ngăn cách. Đây là một dấu hiệu trực quan cho thấy đối tượng đang được thực hiện là một mảng NumPy thay vì Python list.

- Hàm `type()` xác nhận rằng biến `np_array` là một **instance** của lớp `numpy.ndarray`.

- NumPy arrays hỗ trợ các **phép toán theo từng phần tử (element-wise operations)** một cách hiệu quả. VD: chúng ta có thể nhân tất cả phần tử trong mảng với `2` bằng một phép toán duy nhất:

````python
# Perform an element-wise operation
doubled_array = np_array * 2

print("Array after multiplication:")
print(doubled_array)
````

- Đối với danh sách Python thông thường, phép toán tương tự thường phải được thực hiện bằng **vòng lặp (loop)** hoặc **list comprehension**.

- Trong khi đó, NumPy xử lý phép toán này trực tiếp trên toàn bộ mảng thông qua **vector hóa (vectorization)**. Cách tiếp cận này thường ngắn gọn và hiệu quả hơn khi xử lý lượng lớn dữ liệu số.

## **3. Your First Pandas Objects: Series and DataFrame**

- Pandas cung cấp hai **cấu trúc dữ liệu chính (primary data structures**):

  - `Series`: cấu trúc dữ liệu một chiều
  - `DataFrame`: cấu trúc dữ liệu hai chiều



### **3.1. Create a `Series`**

- `Series` có thể được xem như một mảng NumPy một chiều nhưng được bổ sung một **chỉ mục liên kết (associated index)**.

````python
# Tạo dữ liệu từ một danh sách
data_series = [10, 20, 30, 40, 50]

# Tạo Pandas Series từ danh sách
pd_series = pd.Series(data_series)

# In Series
print("My first Pandas Series:")
print(pd_series)

# In kiểu của đối tượng Series
print("\nType of the object:")
print(type(pd_series))
````

- Kết quả hiển thị hai thành phần chính:

  - Các **giá trị dữ liệu (data values)**: `10`, `20`, `30`, `40`, `50`
  - **Chỉ mục (index)** ở phía bên trái: `0`, `1`, `2`, `3`, `4`

- Do chúng ta không cung cấp **index** cụ thể, Pandas tự động tạo **index mặc định** bắt đầu từ `0`.

- Dòng `dtype: int64`: cho biết **kiểu dữ liệu (data type)** của các giá trị được lưu trữ trong `Series` là số nguyên 64 bit

### **3.2. Create a `DataFrame`**

- `DataFrame` là một **cấu trúc dữ liệu dạng bảng hai chiều (two-dimensional tabular data structure)**. Cấu trúc này bao gồm các hàng và cột, trong đó mỗi cột có thể chứa một kiểu dữ liệu khác nhau.
- Một phương pháp phổ biến để tạo `DataFrame` là sử dụng **từ điển Python (Python dictionary)**. Trong cách tổ chức này:
  - Các **key** của từ điển trở thành tên cột.
  - Các **lists** hoặc **arrays** tưng ứng trở thành data của từng cột.

````python
# Tạo một từ điển dữ liệu
data_dict = {
    'StudentID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Score': [85, 92, 78, 88]
}

# Tạo Pandas DataFrame từ từ điển
df = pd.DataFrame(data_dict)

# In DataFrame
print("My first Pandas DataFrame:")
print(df)

# In kiểu của đối tượng DataFrame
print("\nType of the object:")
print(type(df))
````

- Kết quả hiển thị một bảng dữ liệu gồm ba cột:
  - `StudentID`
  - `Name`
  - `Score`

  - Pandas cũng tự động tạo **chỉ mục hàng (row index)** gồm các giá trị: `0, 1, 2, 3`.

- Có thể diễn giải `DataFrame` như sau

| index | StudentID | Name    | Score |
| ------: | --------: | ------- | ----: |
|       0 |       101 | Alice   |    85 |
|       1 |       102 | Bob     |    92 |
|       2 |       103 | Charlie |    78 |
|       3 |       104 | David   |    88 |
