# **Importance for AI and Data Science**

- Các ứng dụng Artificial Intelligence và Data Science đều được xây dựng dựa trên **dữ liệu (data)**. Cho dù mục tiêu là huấn luyện một **mô hình học máy (machine learning model)** để nhận dạng hình ảnh, phân tích hành vi khách hàng hay xử lý dữ liệu từ cảm biến, khả năng xử lý và thao tác dữ liệu một cách hiệu quả luôn là yếu tố cốt lõi.

- Tuy nhiên, **dữ liệu thô (raw data)** hiếm khi ở trạng thái sẵn sàng để phân tích hoặc **huấn luyện (train)** mô hình. Trong hầu hết các trường hợp, dữ liệu cần được **làm sạch dữ liệu (data cleaning)**, **biến đổi dữ liệu (data transformation)** và **tổ chức dữ liệu (data structuring)** trước khi có thể sử dụng.

- Python, với **hệ sinh thái thư viện (library ecosystem)** phong phú, đã trở thành ngôn ngữ lập trình phổ biến nhất trong AI và Data Science. Trong hệ sinh thái này, NumPy và Pandas được xem là hai thư viện nền tảng cho việc xử lý dữ liệu.

## **1. Efficient Numerical Computation with NumPy**

- Phần lớn các bài toán trong DS đều liên quan đến **tính toán số (numerical computation)** trên **những tập dữ liệu lớn (large datasets)**.

- Mặc dù Python lists và loops có tính linh hoạt cao, chúng lại không đủ hiệu quả khi thực hiện các phép tính toán trên khối lượng lớn dữ liệu số.

- NumPy được phát triển để giải quyết vấn đề này, nó cung cấp các thành phần quan trọng:

  - **Đối tượng `ndarray`**: NumPy giới thiệu **đối tượng mảng N chiều (N-dimensional array object)** hay `ndarray`. So với các **kiểu dữ liệu tuần tự (Python sequences)** trong Python, `ndarray` mang lại hai ưu điểm nổi bật:
    - Sử dụng bộ nhớ hiệu quả
    - Thực hiện các phép toán số nhanh hơn đáng kể.

    -> Điều này đạt được qua các phép toán được triển khai dưới dạng **mã C đã biên dịch và tối ưu hóa (optimized pre-compiled C code)**.

  - **Phép toán vector hóa (Vectorized Operations)**:
    - Một trong những đặc điểm quan trọng nhất của NumPy là hỗ trợ **phép toán vector hóa (vectorized operations)**. Thay vì sử dụng vòng lặp Python để xử lý từng phần tử, NumPy cho phép thực hiện phép toán trên toàn bộ mảng chỉ bằng một câu lệnh.

    - Ví dụ, khi cộng hai mảng NumPy, phép cộng sẽ được thực hiện theo từng phần tử tương đương.
    
    - Cơ chế vectorized operations này giúp tăng tốc độ tính toán lên rất nhiều so với việc sử dụng vòng lặp thông thường.

  - **Mathematical Functions**: NumPy cung cấp một thư viện phong phú gồm các **hàm toán học cấp cao (high-level mathematical functions)** phục vụ cho nhiều lĩnh vực khác nhau:
    - **Đại số tuyến tính (Linear Algebra)**
    - **Biến đổi Fourier (Fourier Transform)**
    - **Sinh số ngẫu nhiên (Random Number Generation)**
    - Nhiều phép toán khoa học và thống kê khác.

    -> Phần lớn các **thuật toán học máy (machine learning algorithms)** đều yêu cầu dữ liệu đầu vào ở dạng **mảng số (numerical arrays)**.

- NumPy cung cấp định dạng chuẩn cùng các công cụ cần thiết để thao tác với các mảng này một cách hiệu quả. Vì vậy, NumPy được xem là nền tảng không thể thiếu của **tính toán khoa học (scientific computing)** và AI trong Python.

## **2. Flexible Data Structures and Analysis with Pandas**

- Trong khi NumPy rất mạnh trong việc xử lý các mảng số, data trong thực tế thường có cấu trúc phức tạp hơn. Một số data có thể đồng thời chứa: Số, Văn bản, Ngày tháng, **Dữ liệu phân loại (categorical data)**.

- Các dữ liệu này thường được tổ chức dưới dạng bảng tương tự như **bảng tính (spreadsheet)** hoặc **bảng trong cơ sở dữ liệu (database table)**.

- Pandas được xây dựng trên nền tảng của NumPy nhằm hỗ trợ xử lý chính xác loại dữ liệu này.

  - **DataFrame and Series**: Pandas giới thiệu hai cấu trúc dữ liệu chính:
    - `DataFrame`: **Cấu trúc dữ liệu 2 chiều có nhãn (2-dimensional labeled data structure)**. Có thể hình dung `DataFrame` như một bảng dữ liệu trong đó:
      - Mỗi hàng biểu diễn một bản ghi.
      - Mỗi cột biểu diễn một thuộc tính.

    - `Series`: **Mảng một chiều có nhãn (1-dimensional labeled array)**. Một `Series` có thể được xem như một cột riêng lẻ trong `DataFrame`.


    - Cả `DataFrame` và `Series`: Đều được hỗ trợ truy cập dữ liệu thông qua:
      - Chỉ mục hàng (row index)
      - Nhãn cột (column labels)
    
  - **Data Handling Capabilities**: Pandas đơn giản hóa hầu hết các công việc thường gặp trong quá trình xử lý data. Các khả năng bao gồm:
    - Đọc data từ nhiều định dạng khác nhau: **CSV, Excel, JSON và cơ sở dữ liệu SQL (SQL databases)**.
    - Xử lý **giá trị thiếu (missing values)**:
    - Lọc và lựa chọn data theo điều kiện
    - **Thay đổi cấu trúc data (data reshaping)**.
    - **Hợp nhất nhiều tập data (merging multiple datasets)**.
    - Thực hiện **phân tích theo nhóm (group-wise analysis)**.

  - **Integration**: Một ưu điểm lớn của Pandas là khả năng tích hợp chặt chẽ với nhiều thư viện khác trong hệ sinh thái Python. Ví dụ:
    - **Numpy** cho các phép toán số.
    - **Matplotlib** để **trực quan hóa dữ liệu (data visualization)**.
    - **Scikit-learn** để xây dựng **mô hình học máy (machine learning models)**.
    
- Sau khi dữ liệu được làm sạch và chuẩn bị bằng Pandas, dữ liệu thường có thể được đưa trực tiếp vào các mô hình học máy.

## **3. The Typical Workflow**

- Trong một dự án AI hoặc DS, NumPy và Pandas thường được sử dụng xuyên suốt giai đoạn chuẩn bị dữ liệu:

  - **loading**: Pandas được sử dụng để **nạp data (data loading)** từ nhiều nguồn khác nhau và lưu trữ chúng trong các đối tượng `DataFrame`.

  - **Inspection & Cleaning**: Sau khi data loading, cả NumPy và Pandas đều hỗ trợ quá trình **kiểm tra data (data inspection)**. Trong đó, Pandas đặc biệt hữu ích cho các tác vụ như:
    - Phát hiện và xử lý giá trị thiếu
    - Điều chỉnh **kiểu dữ liệu (data types)**
    - Phát hiện các **điểm không nhất quán (data inconsistencies)**

  - **Transformation & Feature Engineering**:
    - Data thường cần được:
      - **Biến đổi (transformation)**
      - **Tổng hợp (aggregation)**
      - **Kết hợp (combination)**

    - Ngoài ra, nhiều bài toán yêu cầu tạo ra các **đặc trưng mới (feature engineering)** từ dữ liệu hiện có.

    - Pandas cung cấp nhiều công cụ mạnh mẽ để thực hiện các tác vụ này, chẳng hạn như: `groupby`, `merge`, `apply`
    - Trong khi đó, NumPy thường được sử dụng để thực hiện các phép biến đổi mang tính số học.

  - **Preparation for Modeling**: Trước khi đưa data vào các mô hình máy học, trong `DataFrame` thường được chuyển đổi thành **NumPy arrays**. Đây là định dạng đầu vào tiêu chuẩn mà nhiều thư viện học máy, chẳng hạn như **Scikit-learn** yêu cầu:

- Nếu không có các **mảng số hiệu quả (efficient numerical arrays)** của NumPy và các **cấu trúc dữ liệu linh hoạt (flexible data structures)** cùng công cụ thao tác dữ liệu của Pandas, quá trình chuẩn bị data cho các bài toán AI và DS trong Python sẽ trở nên phức tạp hơn rất nhiều, đồng thời hiệu năng xử lý cũng giảm.

- Vì vậy, việc thành thạo NumPy và Pandas là một bước nền tảng quan trọng đối với bất kỳ ai muốn theo đuổi lĩnh vực AI và DS.
