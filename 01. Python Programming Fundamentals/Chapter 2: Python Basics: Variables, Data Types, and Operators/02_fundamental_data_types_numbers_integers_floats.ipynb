# **Fundamental Data Types: Numbers (Integers, Floats)**

- Ở cốt lõi của mọi hoạt động tính toán là khả năng làm việc với con số. Dù bạn đang đếm số lượng vật thể, tính khoảng cách hay đo các thuộc tính vật lý, con số luôn giữ vai trò không thể thiếu. Python cung cấp sẵn sự hỗ trợ cho nhiều kiểu dữ liệu số, nhưng hai kiểu cơ bản nhất mà bạn sẽ gặp thường xuyên là **số nguyên (integers)** và **số thực dấu chấm động (floating-point numbers)**.

## **1. Integers (int)**

- Integers là các số **không có phần thập phân**. Chúng có thể là số dương, số âm hoặc bằng 0. Có thể xem số nguyên như những giá trị dùng để đếm các đối tượng rời rạc, chẳng hạn như số người dùng, số bước trong một vòng lặp, hoặc điểm số trong một trò chơi.

- Trong Python, bạn tạo một số nguyên đơn giản bằng cách viết **con số đó không có dấu chấm thập phân**:

```python
# Examples of integers
user_count = 150
temperature_celsius = -5
year = 2024
zero_value = 0

print(user_count)
print(temperature_celsius)
```

- Integers trong python có thể có giá trị rất lớn, giới hạn duy nhất là dung lượng bộ nhớ của hệ thống. Bạn sẽ không cần quan tâm tới kích thước integer như `short`, `int`, `long` như trong một số ngôn ngữ lập trình khác.

- Bạn có thể kiểm tra kiểu dữ liệu của một số bằng hàm tích hợp sẵn `type()`:

```python
age = 30
print(type(age)) # Output: <class 'int'>
```

## **2. Floating-Point Numbers (float)**

- Floating-point number hoặc floats, đại diện có các số thực, bao gồm cả những giá trị có phần thập phân. Chúng đặc biệt quan trọng khi làm việc với các phép đo, tỷ lệ phần trăm, hoặc bất kỳ giá trị nào cần độ chính xác vượt ra ngoài số nguyên.

- Bạn có thể tạo một số thực bằng cách thêm dấu chấm thập phân hoặc sử dụng ký pháp khoa học (scientific notation, hay e notation).

```python
# Examples of floating-point numbers
pi_approx = 3.14159
price = 49.99
temperature_fahrenheit = -20.5
account_balance = 0.0 # Using a decimal makes it a float
distance_meters = 1.5e3 # Scientific notation for 1.5 * 10^3, which is 1500.0

print(pi_approx)
print(distance_meters)
```
