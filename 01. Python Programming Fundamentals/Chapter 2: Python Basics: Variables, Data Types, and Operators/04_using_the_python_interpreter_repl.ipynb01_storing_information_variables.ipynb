# **Storing Information: Variables**

- Các chương trình luôn làm việc với những mẫu thông tin như **numbers**, **text** hoặc **kết quả trả về từ các phép tính**. Thay vì sử dụng trực tiếp dữ **liệu thô (raw data)** cho chương trình, cách tốt hơn đó là đặt tên cho data đó và tham chiếu đến nó thông qua tên. Đây chính là vai trò **biến (variable)**

- Một **variable** giống như một chiếc hộp có nhãn hoặc một vị trí giữ chỗ có tên trong bộ nhớ máy tính, nơi bạn có thể lưu trữ một giá trị.

## **1. What is a Variable?**

- Trong Python, **variable** là một tên **ký hiệu (symbolic name)** đại diện cho một giá trị. Bạn sử dụng variable để lưu trữ data, sau đó có thể tham chiếu và xử lý data đó trong chương trình
 > `age = 25`: `age` chính là symbolic name, tên tượng trưng mà bạn đặt cho địa chỉ bộ nhớ `0x001`.

- Thuật ngữ variable hàm ý rằng **giá trị được lưu trữ có thể thay đổi** trong quá trình chương trình thực thi.

- Hãy tưởng tượng bạn đang sắp xếp đồ vật vào các chiếc hộp. Bạn sẽ không muốn đặt mọi thứ vào những chiếc hộp không có nhãn; thay vào đó, bạn sẽ dán nhãn cho chúng như `“Quần áo mùa đông”`, `“Sách”`, hoặc `“Dụng cụ nhà bếp”`. Variable trong lập trình cũng có chức năng tương tự: chúng cung cấp tên mô tả cho các vị trí bộ nhớ đang chứa dữ liệu.

## **2. Creating Variables: The Assignment Statement**

- Việc tạo một variable trong Python rất đơn giản bằng cách sử dụng **toán tử gán** (`=`). Cấu trúc có dạng:

    ```python
    variable_name = value
    ```
- Cách hoạt động như sau:

  1. Bạn chọn một tên cho variable của mình (tuân theo một số quy tắc).
  2. Bạn sử dụng dấu =.
  3. Bạn cung cấp giá trị mà bạn muốn lưu vào variable.

````python
message = "Hello, Python Learner!"
count = 10
price = 49.99
is_active = True
````

- Trong các ví dụ này:
  - `message` là một variable lưu trữ một chuỗi văn bản.
  - `count` là một variable lưu trữ một số integer.
  - `price` là một variable lưu trữ một floating-point number.
  - `is_active` là một variable lưu trữ giá trị logic (boolean) `True`.

- Python là một **ngôn ngữ định kiểu động (dynamically typed)**, nghĩa là bạn không cần khai báo tường minh kiểu dữ liệu mà variable sẽ chứa. Python sẽ tự động xác định kiểu dựa trên giá trị mà bạn gán cho nó.

- Khi bạn gán `"Hello, Python Learner!"` cho `message`, Python hiểu rằng `message` lúc này đang chứa một string.

## **3. Using Variables**

- Sau khi đã tạo (hoặc khai báo và khởi tạo) một variable, bạn có thể sử dụng tên của nó trong mã nguồn để truy cập giá trị mà nó đang giữ.

````python
message = ""
print(message) # Output:

items = 5
cost_per_item = 2.50
total_cost = items * cost_per_item # Use variables in calculations
print(total_cost) # Output: 12.5
````

- Việc sử dụng variable giúp mã nguồn dễ đọc hơn và dễ bảo trì hơn.

- Nếu `cost_per_item` thay đổi, bạn chỉ cần cập nhật giá trị ở một nơi duy nhất, tức chỗ variable được gán, thay vì phải tìm tất cả các lần xuất hiện của số 2.50 trong toàn bộ mã.

## **4. Naming Variables: Rules and Conventions**

- Việc lựa chọn tên variable tốt là rất quan trọng để viết mã rõ ràng và dễ hiểu. Python có các quy tắc cụ thể và những quy ước được chấp nhận rộng rãi khi đặt tên variable.

### **4.1 Rules (Must be followed)**

1. Tên variable phải bắt đầu bằng một chữ cái (`a-z`, `A-Z`) hoặc dấu gạch dưới (`_`). Tên variable không được bắt đầu bằng chữ số.

2. Phần còn lại của tên có thể bao gồm chữ cái, chữ số (`0-9`) và dấu gạch dưới.

3. Tên variable phân biệt hoa thường. `myVariable`, `myvariable` và `MYVARIABLE` là ba variable khác nhau.

4. Tên variable không được trùng với **từ khóa của Python**, chẳng hạn như `if, else, for, while, def, class, True, False, None`.

### **4.2 Conventions (Should be followed for readability)**

- Dùng tên mô tả rõ mục đích của variable, ví dụ `user_name` thay vì `un`, hoặc `total_price` thay vì `tp`.

- Dùng kiểu **snake_case** cho tên variable gồm nhiều từ. Điều này có nghĩa là tất cả chữ cái đều viết thường, và các từ được ngăn cách bằng dấu gạch dưới, ví dụ `first_name`, `number_of_items`.

=> Một số ví dụ về tên variable hợp lệ và không hợp lệ:

- **Hợp lệ**: `name, age, user_input, _internal_counter, price_2024`
- **Không hợp lệ**: `2nd_place` (bắt đầu bằng số), `user-name` (chứa dấu gạch ngang), class (là từ khóa)

## **5. Changing Variable Values (Reassignment)**

- Bạn có thể thay đổi giá trị được lưu trong một variable bằng cách gán cho nó một giá trị mới bằng toán tử `=` một lần nữa.

````python
# Assign an initial value
score = 100
print("Initial score:", score) # Output: Initial score: 100

# Update the value
score = 150
print("Updated score:", score) # Output: Updated score: 150

# Assign a value based on the current value
score = score + 20 # Read the current value (150), add 20, store the result (170) back
print("Final score:", score)   # Output: Final score: 170
````

- Mỗi khi bạn gán một giá trị mới, giá trị cũ gắn với tên variable đó sẽ được thay thế.

variable là một trong những khối xây dựng nền tảng của Python. Chúng cho phép bạn đặt tên và lưu trữ dữ liệu, từ đó làm cho chương trình trở nên linh hoạt và dễ hiểu hơn. Khi bạn tiến xa hơn, bạn sẽ thấy variable được sử dụng liên tục để quản lý luồng thông tin bên trong mã nguồn của mình.
