# **Getting User Input**

- Khi chương trình gọi `input()`, quá trình thực thi sẽ **tạm dừng** và chờ người dùng nhập dữ liệu vào console rồi nhấn **Enter**. Nội dung mà người dùng nhập sẽ được hàm này **trả về dưới dạng một string**.

```python
print("What is your name?")
user_name = input()
print("Hello,", user_name)
```

- Khi bạn chạy đoạn này, chương trình sẽ in ra `"What is your name?"`, sau đó chờ. Nếu bạn nhập `Alice` rồi nhấn Enter, chương trình sẽ tiếp tục và in ra `Hello, Alice`. Văn bản bạn nhập vào: `Alice`, đã được `input()` lấy và lưu vào biến `user_name`.

## **1. Providing a Prompt**

- Thông thường, sẽ thân thiện với người dùng hơn nếu bạn đặt luôn một **prompt message** ngay trong hàm `input()`. Thông báo này sẽ hiển thị cho người dùng biết họ cần nhập thông tin gì.

```python
user_name = input("Enter your name: ")
print("Hello,", user_name)
```

- Phiên bản này cho kết quả tương tự như ví dụ trước, nhưng gộp prompt và yêu cầu nhập vào cùng một dòng, giúp code ngắn gọn hơn một chút. Chuỗi `"Enter your name: "` sẽ được hiển thị, và chương trình sẽ chờ nhập ngay sau đó.

## **2 `input()` is Always a String**

- Một điểm rất quan trọng của hàm `input()` là nó luôn luôn trả về dữ liệu người dùng nhập dưới dạng **string**, bất kể người dùng nhập gì. Ngay cả khi người dùng nhập các chữ số, `input()` vẫn xem chúng là các ký tự tạo thành một **string**.

```python
age_input = input("Enter your age: ")
print("Type of age_input:", type(age_input))
# Attempting addition might cause an error or unexpected behavior
# next_year = age_input + 1 # This would raise a TypeError
```

- Lưu ý rằng kiểu dữ liệu là `<class 'str'>`, không phải `<class 'int'>`. Vì `age_input` là một string `("25")`, bạn không thể trực tiếp thực hiện phép toán với nó như cộng `1`. Nếu cố tính `age_input + 1`, chương trình sẽ báo lỗi `TypeError`, vì Python không biết cách cộng một integer với một string trong ngữ cảnh này.

## **3. Converting Input Types**

- Nếu bạn cần xử lý dữ liệu người dùng như một số nguyên hoặc số thực, bạn phải **chủ động chuyển đổi** string do `input()` trả về. Bạn có thể dùng các hàm chuyển kiểu đã học như `int()` hoặc `float()`

```python
age_str = input("Enter your age: ")
age_int = int(age_str) # Convert the input string to an integer

next_year_age = age_int + 1
print("Next year, you will be", next_year_age)

# You can also combine input and conversion in one step:
height_str = input("Enter your height in meters: ")
height_float = float(height_str) # Convert the input string to a float

print("Your height is:", height_float, "meters")
```

- Trong ví dụ này:

    - Chúng ta lấy tuổi dưới dạng string bằng `input()` và lưu vào `age_str`.
    - Chúng ta dùng `int(age_str)` để chuyển string `"25"` (hoặc bất kỳ giá trị nào người dùng nhập) thành integer `25`, rồi lưu vào `age_int`.
    - Bây giờ ta có thể thực hiện phép toán: `age_int + 1`.
    - Tương tự, ta chuyển input về chiều cao thành **floating-point number** bằng `float()` để có thể xử lý các giá trị thập phân.
    
- Hãy nhớ rằng nếu người dùng nhập text không thể chuyển sang kiểu đích (ví dụ nhập `"hello"` khi dùng `int()`), chương trình sẽ báo lỗi `ValueError`. Cách xử lý các lỗi như vậy sẽ được đề cập sau, trong phần nói về exception handling.
