# **Working with Operators: Logical**

- Các toán tử như `== (bằng nhau)`, `!= (không bằng)`, `< (nhỏ hơn)`, và `> (lớn hơn)` dùng để so sánh các giá trị và luôn cho ra một kết quả boolean: `True` hoặc `False`. Các **toán tử logic (logical operators)** cho phép kết hợp những kết quả boolean này để tạo ra các biểu thức điều kiện phức tạp hơn.

- Trong thực tế, bạn thường cần kiểm tra nhiều điều kiện cùng lúc. Ví dụ, bạn có thể muốn biết liệu một người dùng vừa đăng nhập vừa có quyền quản trị hay không, hoặc hôm nay có phải thứ Bảy hoặc Chủ nhật hay không. Đây chính là lúc các toán tử logic phát huy tác dụng. Chúng cho phép bạn kết hợp các giá trị boolean (`True` hoặc `False`) để tạo thành những điều kiện phức tạp hơn. Python cung cấp ba toán tử logic:
- `and`
- `or`
- `not`

## **1. The `and` Operator**

- Toán tử `and` đánh giá hai biểu thức boolean. Nó chỉ trả về `Tru`e khi cả hai biểu thức đều là `True`. Nếu một trong hai biểu thức (hoặc cả hai) là `False`, kết quả sẽ là `False`.

- Hãy tưởng tượng bạn cần hai loại giấy tờ để vào một khu vực an ninh cao. Bạn cần cả thẻ A và thẻ B. Chỉ có một trong hai là chưa đủ.

```python
age = 25
has_ticket = True

# Người này có đủ tuổi VÀ có vé không?
can_enter = (age >= 18) and (has_ticket == True)
print(f"Age is 25, has ticket: {can_enter}") # Kết quả: True

age = 16
# Bây giờ điều kiện đầu tiên (age >= 18) là False
can_enter = (age >= 18) and (has_ticket == True)
print(f"Age is 16, has ticket: {can_enter}") # Kết quả: False

age = 30
has_ticket = False
# Bây giờ điều kiện thứ hai (has_ticket == True) là False
can_enter = (age >= 18) and (has_ticket == True)
print(f"Age is 30, no ticket: {can_enter}") # Kết quả: False
```

- Các dấu `()` xung quanh biểu thức so sánh như (`age >= 18`) không bắt buộc trong trường hợp này vì `and` có **độ ưu tiên (precedence)** thấp hơn các **comparison operators**. Tuy nhiên, việc dùng ngoặc giúp mã dễ đọc hơn bằng cách nhóm rõ các điều kiện.

## **2. The `or` Operator**

- Toán tử `or` cũng đánh giá hai biểu thức boolean. Nó trả về `True` nếu ít nhất một biểu thức là `True`. Nó chỉ trả về `False` khi cả hai biểu thức đều là False.

- Hãy tưởng tượng bạn được giảm giá nếu bạn là sinh viên hoặc người cao tuổi. Chỉ cần thỏa mãn một trong hai điều kiện là đủ.

```python
is_weekend = False
is_holiday = True

# Hôm nay là cuối tuần HOẶC ngày nghỉ lễ?
day_off = is_weekend or is_holiday
print(f"Weekend: False, Holiday: True -> Day off: {day_off}") # Kết quả: True

is_weekend = True
is_holiday = False
# Điều kiện đầu tiên là True
day_off = is_weekend or is_holiday
print(f"Weekend: True, Holiday: False -> Day off: {day_off}") # Kết quả: True

is_weekend = False
is_holiday = False
# Cả hai điều kiện đều False
day_off = is_weekend or is_holiday
print(f"Weekend: False, Holiday: False -> Day off: {day_off}") # Kết quả: False
```

## **3. The `not` Operator**

- Toán tử `not` đơn giản hơn. Nó hoạt động trên một biểu thức boolean duy nhất và đảo ngược giá trị của biểu thức đó.

- Nếu biểu thức là `True`, `not` sẽ biến nó thành `False`. Nếu biểu thức là `False`, `not` sẽ biến nó thành `True`.

```python
is_logged_in = False
print(f"Is logged in: {is_logged_in}") # Kết quả: False

# Kiểm tra xem người dùng CHƯA đăng nhập
needs_login = not is_logged_in
print(f"Needs login: {needs_login}") # Kết quả: True

is_raining = True
print(f"Is raining: {is_raining}") # Kết quả: True

# Kiểm tra xem trời KHÔNG mưa
can_go_outside = not is_raining
print(f"Can go outside: {can_go_outside}") # Kết quả: False
```

## **4. Combining Logical Operators**

- Bạn có thể kết hợp nhiều toán tử logic lại trong một biểu thức.
- Thứ tự ưu tiên của chúng trong Python: `not` - `and` - `or`
- Tuy nhiên, bạn có thể dùng dấu ngoặc `()` để:
  - Kiểm soát thứ tự, và làm code dễ đọc hơn

```python
age = 22
is_student = True
has_coupon = False

# Được giảm giá nếu (tuổi < 25 VÀ là sinh viên) HOẶC có mã giảm giá
is_eligible_discount = (age < 25 and is_student) or has_coupon

# Được đánh giá như sau:
# (True and True) or False
# -> True or False
# -> True

print(f"Eligible for discount: {is_eligible_discount}") # Kết quả: True

# Thay đổi trạng thái mã giảm giá
has_coupon = True
age = 30
is_student = False

is_eligible_discount = (age < 25 and is_student) or has_coupon

# Được đánh giá như sau:
# (False and False) or True
# -> False or True
# -> True

print(f"Eligible for discount (different criteria): {is_eligible_discount}") # Kết quả: True
```
