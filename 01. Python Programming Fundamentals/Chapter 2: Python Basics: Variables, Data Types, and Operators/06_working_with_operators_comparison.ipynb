# **Working with Operators: Comparison**

- Trong lập trình, chúng ta thường cần so sánh các giá trị để đưa ra quyết định. Giống như cách các **toán tử số học (arithmetic operators)** như `+` và `-` được dùng để thực hiện phép tính.

- Python cũng cung cấp các **toán tử so sánh (comparison operators)**, còn được gọi là **relational operators**, để kiểm tra mối quan hệ giữa hai giá trị. Những phép so sánh này là nền tảng để điều khiển luồng chạy của chương trình, cho phép chương trình phản ứng khác nhau dựa trên các điều kiện cụ thể.

- Kết quả của mọi phép so sánh luôn là một **giá trị boolean (Boolean value)**: hoặc `True`, hoặc `False`.

## **1. Equal To: `==`**
- Toán tử này kiểm tra xem giá trị ở bên trái và bên phải có bằng nhau hay không.

- Điều quan trọng là phải phân biệt **toán tử so sánh (comparison operato `==`)** với **toán tử gán (assignment operator `=`)**, vốn được dùng để gán giá trị cho biến.

- Việc dùng một dấu `=` trong khi muốn so sánh là lỗi rất phổ biến đối với người mới học lập trình.

```python
# So sánh số
print(5 == 5)
print(10 == 7)

# So sánh chuỗi
print("hello" == "hello")
print("Python" == "python") # Phân biệt chữ hoa chữ thường!

# Gán kết quả vào biến
are_numbers_equal = (100 == 100)
print(are_numbers_equal)
```

## **2. Not Equal To: `!=`**

- Toán tử này kiểm tra xem giá trị ở bên trái và bên phải có khác nhau hay không. Đây là phép ngược trực tiếp của toán tử `==`.

```python
# So sánh số
print(5 != 5)
print(10 != 7)

# So sánh chuỗi
print("hello" != "")
print("Python" != "Python")

# Gán kết quả
are_different = ("apple" != "orange")
print(are_different)
```

## **3. Less Than: `<`**

- Toán tử này kiểm tra xem giá trị bên trái có nhỏ hơn một cách nghiêm ngặt giá trị bên phải hay không.

```python
print(5 < 10)
print(10 < 5)
print(5 < 5) # Không nhỏ hơn, nên là False

# Chuỗi được so sánh theo thứ tự từ điển (lexicographically)
print("apple" < "banana")
print("cat" < "car") # 't' đứng sau 'r'
```

## **4. Greater Than: `>`**

- Toán tử này kiểm tra xem giá trị bên trái có lớn hơn một cách nghiêm ngặt giá trị bên phải hay không.

```python
print(10 > 5)
print(5 > 10)
print(5 > 5) # Không lớn hơn, nên là False

print("zebra" > "apple")
```

## **5. Less Than or Equal To: `<=`**

- Toán tử này kiểm tra xem giá trị bên trái có nhỏ hơn hoặc bằng giá trị bên phải hay không.

```python
print(5 <= 10)
print(10 <= 5)
print(5 <= 5) # Bằng nhau, nên là True
```

## **6. Greater Than or Equal To: `>=`**

- Toán tử này kiểm tra xem giá trị bên trái có lớn hơn hoặc bằng giá trị bên phải hay không.

```python
print(10 >= 5)
print(5 >= 10)
print(5 >= 5) # Bằng nhau, nên là True
```

## **7. Comparing Different Numeric Types**

- Python linh hoạt khi so sánh các kiểu số khác nhau, như **số nguyên (integer)** và **số thực (float)**. Thông thường, Python sẽ chuyển chúng về một kiểu chung trước khi so sánh.



```python
print(5 == 5.0)
print(10 > 9.99)
print(3 <= 3.0)
```

## **8. Summary of Comparison Operators**

- Dưới đây là bảng tóm tắt nhanh:

| Operator | Meaning              | Example  | Result |
|:---------|:---------------------|:---------|:-------|
| `==`     | Equal to             | `5 == 5` | `True`   |
| `!=`     | Not equal to         | `5 != 6` | `True`   |
| `<`      | Less than            | `5 < 10` | `True`   |
| `>`      | Greater than         | `10 > 5` | `True`   |
| `<=`     | Less than or equal to | `5 <= 5` | `True`   |
| `>=`     | Greater than or equal to | `10 >= 5`| `True`   |

- Hiểu các toán tử này là rất quan trọng vì chúng tạo nên nền tảng của việc ra quyết định trong lập trình. Ở chương tiếp theo, **Kiểm soát luồng chương trình**, bạn sẽ thấy kết quả `True` hoặc `False` từ các phép so sánh này được sử dụng như thế nào trong các câu lệnh `if` và **vòng lặp (loops)** để điều khiển cách chương trình thực thi.
