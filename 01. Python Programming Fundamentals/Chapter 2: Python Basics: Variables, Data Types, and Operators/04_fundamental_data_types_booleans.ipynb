# **Fundamental Data Types: Booleans**

- Python xử lý nhiều loại dữ liệu khác nhau, bao gồm số (như `42` và `3.14`) và văn bản (như `"Hello, Python"`).

- Tuy nhiên, trong lập trình, chúng ta thường cần biểu diễn những khái niệm đơn giản hơn như điều kiện đúng hoặc sai, một điều kiện có được thỏa mãn hay không, hoặc một trạng thái nào đó có tồn tại hay không. Trong những trường hợp này, Python cung cấp kiểu dữ liệu **boolean**.

## **1. The Boolean Type: `True` and `False`**

- Kiểu boolean được đặt theo tên của George Boole — người đã phát triển đại số Boolean. Đây là nền tảng quan trọng trong logic lập trình.

- Kiểu dữ liệu này chỉ có hai giá trị duy nhất:
  - `True`
  - `False`

- Hãy chú ý cách viết hoa. `True` và `False` là các **từ khóa đặc biệt (reserved keywords)** trong Python, đại diện cho các giá trị boolean. Bạn không thể dùng `true` hoặc `false` viết thường.

```python
is_learning = True
is_finished = False

print(is_learning)
print(is_finished)
```

- Các biến `is_learning` và `is_finished` hiện đang lưu trữ giá trị boolean. Chúng hoạt động như những “switches” hoặc “flags” đơn giản trong chương trình.

## **2. Booleans from Comparisons**

- Mặc dù bạn có thể gán trực tiếp `True` hoặc `False`, nhưng trong thực tế, giá trị boolean thường xuất hiện như kết quả của các phép so sánh.

- Khi sử dụng các toán tử so sánh như:

  - `==` (bằng)
  - `!=` (khác)
  - `> `(lớn hơn)
  - `< `(nhỏ hơn)
  - `>=` (lớn hơn hoặc bằng)
  - `<=` (nhỏ hơn hoặc bằng)

- kết quả trả về luôn là một giá trị boolean.

```python
age = 20
is_adult = age >= 18  # age có lớn hơn hoặc bằng 18 không?
print(is_adult)       # Output: True

temperature = 15.5
is_cold = temperature < 10.0 # nhiệt độ có nhỏ hơn 10.0 không?
print(is_cold)               # Output: False

name = "Alice"
is_bob = name == "Bob" # tên có bằng "Bob" không?
print(is_bob)          # Output: False
```

- Trong mỗi trường hợp, biểu thức bên phải dấu `=` sẽ được tính toán trước. Phép so sánh sẽ tạo ra `True` hoặc `False`, sau đó kết quả được lưu vào biến.

- Mối liên hệ trực tiếp giữa phép so sánh và giá trị boolean chính là nền tảng của việc điều khiển hành vi chương trình.

## **3. Booleans and Logical Operations**

- Giá trị boolean là nền tảng cho việc đưa ra quyết định trong Python. Chúng thường được sử dụng cùng các toán tử logic:
  - `and`
  - `or`
  - `not`
- Ví dụ, bạn có thể muốn kiểm tra xem một người dùng vừa đăng nhập vừa có quyền quản trị hay không.

```python
logged_in = True
is_admin = False

# 'and' yêu cầu cả hai vế đều True
can_access_admin_panel = logged_in and is_admin
print(can_access_admin_panel) # Output: False

# 'or' chỉ cần một vế True
can_view_content = logged_in or is_admin
print(can_view_content)       # Output: True

# 'not' đảo ngược giá trị boolean
is_guest = not logged_in
print(is_guest)               # Output: False
```

## **4. Truthiness: When Other Types Act Like Booleans**

- Python có một khái niệm rất hữu ích gọi là **truthiness**.

- Trong các ngữ cảnh yêu cầu giá trị **boolean** (ví dụ trong câu lệnh `if`), Python có thể đánh giá các kiểu dữ liệu khác thành:

  - **truthy** → hoạt động như `True`
  - **falsy** → hoạt động như `False`

- Các giá trị sau được xem là **falsy** trong Python:
  - Chính giá trị `False`
  - Giá trị đặc biệt `None`
  - Số `0` dưới mọi dạng (`0`, `0.0`)
  - Các collection rỗng:
    - Chuỗi rỗng: `""`
    - Danh sách rỗng: `[]`
    - Tuple rỗng: `()`
    - Dictionary rỗng: `{}`
    - Set rỗng: `set()`
- Gần như mọi giá trị khác đều là **truthy**, bao gồm:

  - Số khác 0 - **non-zero numbers**
  - Chuỗi không rỗng - **non-empty strings**
  - Danh sách có phần tử
  - v.v...

- Điều này hữu ích vì nó cho phép viết các kiểm tra ngắn gọn. Ví dụ, thay vì kiểm tra một danh sách có nhiều hơn 0 phần tử: `if len(my_list) > 0:`,
bạn thường chỉ cần viết `if my_list`: vì một danh sách không rỗng là **truthy**, trong khi danh sách rỗng là **falsy**.

```python
items = []
if items: # items rỗng, nên là falsy
    print("Có phần tử trong danh sách.")
else:
    print("Danh sách rỗng.") # Dòng này sẽ được in ra

user_name = "Charlie"
if user_name: # user_name không rỗng, nên là truthy
    print(f"Xin chào, {user_name}!") # Dòng này sẽ được in ra
else:
    print("Thiếu tên người dùng.")

count = 0
if count: # count bằng 0, nên là falsy
    print("Số lượng là số dương.")
else:
    print("Số lượng bằng 0.") # Dòng này sẽ được in ra
```

- Mặc dù `True` và `False` là các giá trị boolean tường minh, việc hiểu **truthiness** giúp bạn diễn giải cách Python đưa ra quyết định trong logic điều kiện. Cơ chế đánh giá boolean ngầm định này khiến mã Python thường trở nên dễ đọc và gọn hơn.

- Boolean tuy đơn giản nhưng vô cùng quan trọng. Chúng cho phép chương trình biểu diễn trạng thái (`True/False`), đánh giá điều kiện thông qua các phép so sánh và toán tử logic, và cuối cùng là đưa ra quyết định — hình thành nên một phần cốt lõi của logic lập trình.
