# **Fundamental Data Types: Text (Strings)**

- Làm việc với text là một phần nền tảng của lập trình. Trong Python, dữ liệu văn bản được biểu diễn bằng **strings**. Một string là một **ordered sequence of characters**. Bất cứ thứ gì bạn đặt trong dấu nháy — chữ cái, số, ký hiệu, khoảng trắng — đều trở thành một string.

## **1. Creating Strings**

- Python khá linh hoạt trong cách định nghĩa string. Bạn có thể dùng **single quotes (`'`)** hoặc **double quotes (`"`)**. Sự linh hoạt này rất hữu ích khi trong string đã có sẵn dấu nháy.

```python
# Using single quotes
message_single = 'Hello, Python learner!'
print(message_single)

# Using double quotes
message_double = "This also works just fine."
print(message_double)

# What if your string contains a single quote? Use double quotes to enclose it.
quote = "He said, 'Python is fun!'"
print(quote)

# What if your string contains a double quote? Use single quotes.
reply = 'She responded, "Indeed it is."'
print(reply)
```

- Nếu bạn cần một string trải dài trên nhiều dòng, hoặc chứa cả dấu nháy đơn và dấu nháy kép mà không cần xử lý phức tạp, Python cung cấp **triple quotes**, gồm `'''` hoặc `"""`.

```python
# Triple double quotes for multi-line string
multi_line_doc = """This is the first line.
This is the second line.
And "quotes" or 'apostrophes' can be used freely inside."""
print(multi_line_doc)

# Triple single quotes work the same way
multi_line_alt = '''Another way to write
multi-line text,
very convenient.'''
print(multi_line_alt)
```

## **2. Basic String Operations**

- Bạn có thể kết hợp các string bằng toán tử `+`, thao tác này gọi là **concatenation**. Bạn cũng có thể lặp lại string bằng toán tử `*`.

```python
first_name = "Ada"
last_name = "Lovelace"
space = " "

# Concatenation
full_name = first_name + space + last_name
print(full_name) # Output: Ada Lovelace

# Repetition
separator = "-" * 10 # Repeat the hyphen 10 times
print(separator)   # Output: ----------
print(first_name * 3) # Output: AdaAdaAda
```

> **Lưu ý**: Rằng bạn chỉ có thể **concatenate string** với **string** khác.
>  - Nếu cố cộng **string với number trực tiếp**, chương trình sẽ báo lỗi.
>  - Bạn cần chuyển **number** sang **string** trước bằng `str()`, phần này sẽ được học sau trong **Type Conversion**.

## **3. Accessing Characters: Indexing**

- Vì string là một **sequence**, bạn có thể truy cập từng ký tự bằng vị trí của nó, tức là **index**. Python dùng **zero-based indexing**, nghĩa là ký tự đầu tiên có index là `0`, ký tự thứ hai là `1`, và cứ tiếp tục như vậy.

```python
language = "Python"

# Accessing characters by positive index
print(language[0]) # Output: P (the first character)
print(language[1]) # Output: y (the second character)
print(language[5]) # Output: n (the sixth character)

# Accessing characters using negative index (starts from the end)
print(language[-1]) # Output: n (the last character)
print(language[-2]) # Output: o (the second-to-last character)
```

## **4. Extracting Substrings: Slicing**

- Nếu bạn cần nhiều hơn một ký tự, bạn có thể lấy một phần của string, gọi là **substring** hoặc **slice**. Slicing dùng cú pháp `[start:stop:step]`.

  - `start`: vị trí bắt đầu của slice (inclusive).
  - `stop`: vị trí kết thúc của slice (exclusive).
  - `step`: bước nhảy.

```python
language = "Python"

# Get characters from index 1 up to (but not including) index 4
print(language[1:4]) # Output: yth

# Get characters from the beginning up to index 3 (exclusive)
print(language[:3])  # Output: Pyt

# Get characters from index 2 to the end
print(language[2:])  # Output: thon

# Get the whole string (less common, but shows the defaults)
print(language[:])   # Output: Python

# Get every second character
print(language[::2]) # Output: Pto

# Get the string in reverse
print(language[::-1]) # Output: nohtyP
```

- Slicing luôn tạo ra một string mới và không gây **IndexError** nếu index vượt phạm vi; nó chỉ trả về phần string nằm trong khoảng hợp lệ.

## **5. String Immutability**

- Một đặc điểm quan trọng của string trong Python là chúng **immutable**. Điều này có nghĩa là sau khi string được tạo ra, nó không thể bị thay đổi trực tiếp.

```python
greeting = "hello"
# greeting[0] = "H" # This will cause a TypeError: 'str' object does not support item assignment

# To "change" the greeting, create a new string
new_greeting = "H" + greeting[1:]
print(new_greeting) # Output: Hello
print(greeting)     # Output: hello (original string is unchanged)
```

- Tính bất biến này có thể khiến bạn cảm thấy hơi hạn chế, nhưng nó làm cho string trở nên ổn định và an toàn khi dùng trong nhiều ngữ cảnh, ví dụ như làm **dictionary keys**.

## **6. Common String Methods**

- String có nhiều **built-in methods** hỗ trợ các thao tác phổ biến. **Methods** giống như functions, nhưng được gọi trực tiếp trên một object bằng **dot notation** (`.`). Một số method hữu ích:

  - `len(string)`: tuy technically là built-in function chứ không phải method, nhưng rất quan trọng để lấy số ký tự trong string.
  - `.lower()` / `.upper()`: trả về string mới với toàn bộ ký tự chuyển thành chữ thường hoặc chữ hoa.
  - `.strip()` / `.lstrip()` / `.rstrip()`: loại bỏ khoảng trắng ở đầu và/hoặc cuối string.
  - `.find(substring)`: trả về index bắt đầu của lần xuất hiện đầu tiên của substring. Trả về -1 nếu không tìm thấy.
  - `.replace(old, new)`: trả về string mới, thay tất cả old bằng new.
  - `.startswith(prefix)` / .endswith(suffix): trả về True hoặc False tùy string có bắt đầu hoặc kết thúc bằng substring đã cho hay không.
  - `.split(separator)`: trả về một list các substring, tách tại vị trí separator.

```python
text = "  Learning Python is Fun!  "

# Length
print(len(text)) # Output: 27

# Case conversion
print(text.lower()) # Output:   learning python is fun!
print(text.upper()) # Output:   LEARNING PYTHON IS FUN!

# Stripping whitespace
print(text.strip())  # Output: Learning Python is Fun!
print(text.lstrip()) # Output: Learning Python is Fun!
print(text.rstrip()) # Output:   Learning Python is Fun!

# Finding substrings
print(text.find("Python")) # Output: 11 (index where 'Python' starts)
print(text.find("Java"))   # Output: -1 (not found)

# Replacing substrings
print(text.replace("Fun", "Awesome")) # Output:   Learning Python is Awesome!

# Checking start/end (use strip first for accuracy here)
clean_text = text.strip()
print(clean_text.startswith("Learning")) # Output: True
print(clean_text.endswith("!"))        # Output: True

# Splitting a string into a list
words = clean_text.split(" ")
print(words) # Output: ['Learning', 'Python', 'is', 'Fun!']

# Joining a list into a string
joined_string = "---".join(words)
print(joined_string) # Output: Learning---Python---is---Fun!

# Original string remains unchanged due to immutability
print(text) # Output:   Learning Python is Fun!
```

## **7. Formatting Strings**

- Trong nhiều trường hợp, bạn cần tạo string có chứa giá trị của biến. Bạn có thể dùng concatenation (`+`), nhưng cách này nhanh chóng trở nên rườm rà và dễ sai, nhất là khi làm việc với non-string types. Python có các cách formatting tốt hơn.

- Phương pháp hiện đại và thường được ưu tiên là **f-strings (formatted string literals)**, được giới thiệu từ Python 3.6. Bạn tạo f-string bằng cách thêm tiền tố `f` hoặc `F` trước string literal. Bên trong string, bạn có thể đặt biến hoặc expression trực tiếp trong dấu `{}`.

```python
name = "Alice"
age = 30
city = "New York"

# Using f-string for clean formatting
intro_fstring = f"My name is {name}, I am {age} years old, and I live in {city}."
print(intro_fstring)
# Output: My name is Alice, I am 30 years old, and I live in New York.

# You can also include expressions inside the braces
radius = 5
area = 3.14159 * radius**2
print(f"A circle with radius {radius} has an area of {area:.2f}.")
# Output: A circle with radius 5 has an area of 31.42.
# Note: :.2f formats the float to 2 decimal places.
```

- Strings là nền tảng để xử lý mọi dạng dữ liệu văn bản, từ user input, nội dung file, đến messages và labels trong chương trình. Việc nắm vững cách tạo, thao tác và formatting string là một bước rất quan trọng trong quá trình học Python.
