# **Working with Operators: Arithmetic**

- Lập trình Python liên quan đến việc làm việc với dữ liệu số, có thể được lưu trong variables dưới dạng **integer** (như `10`, `-5`, `0`) hoặc **floating-point number** (như `3.14`, `-0.5`, `2.0`).

- Để thực hiện các phép tính với những số này, Python cung cấp một bộ **toán tử số học (arithmetic operators)**. Đây là những công cụ nền tảng, giống như một chiếc calculator, dùng để thực hiện các tác vụ toán học trong chương trình của bạn.

## **1. Basic Arithmetic**

- Python hỗ trợ những phép toán tiêu chuẩn mà bạn đã quen thuộc:

### **1.1. Addition (`+`)**

- Phép cộng dùng để cộng hai số lại với nhau.

```python
apples = 5
oranges = 10
total_fruit = apples + oranges
print(total_fruit) # Kết quả: 15

price = 9.99
tax = 0.80
total_cost = price + tax
print(total_cost) # Kết quả: 10.79
```

### **1.2. Subtraction (`-`)**

- Phép trừ lấy số thứ hai trừ khỏi số thứ nhất.

```python
budget = 100
spent = 35.50
remaining = budget - spent
print(remaining) # Kết quả: 64.5
```

### **1.3. Multiplication (`*`)**

- Phép nhân nhân hai số với nhau.

```python
width = 8
height = 12
area = width * height
print(area) # Kết quả: 96
```

### **1.4. Division (`/`)**

- Phép chia lấy số thứ nhất chia cho số thứ hai. Điều quan trọng là phép chia chuẩn trong Python luôn trả về một **số thực (float)**, ngay cả khi kết quả là một **integer**.

```python
total_distance = 100 # kilômét
time_hours = 2 # giờ
speed = total_distance / time_hours
print(speed) # Kết quả: 50.0

items = 10
people = 4
items_per_person = items / people
print(items_per_person) # Kết quả: 2.5
```

- Hãy chú ý rằng `100 / 2` cho ra `50.0`, chứ không phải chỉ `50`. Cách hoạt động nhất quán này giúp tránh những lỗi tinh vi khi bạn mong đợi kết quả có phần thập phân.

## **2. Integer Division and Remainders**

- Python cung cấp thêm hai toán tử chia chuyên biệt, đặc biệt hữu ích khi làm việc với **số nguyên (integers)**:

### **2.1 Floor Division (`//`)**

- Phép chia lấy phần nguyên hoạt động giống như phép chia `/`, nhưng nó bỏ phần thập phân và làm tròn xuống đến số nguyên gần nhất.
- Đây còn được gọi là **integer division**.

```python
cookies = 10
friends = 3
cookies_per_friend = cookies // friends
print(cookies_per_friend) # Kết quả: 3 (Mỗi người bạn nhận 3 cái bánh nguyên)

value = 7
divisor = 2
result = value // divisor
print(result) # Kết quả: 3 (7 chia 2 là 3.5, làm tròn xuống còn 3)

# So sánh với phép chia thông thường
print(10 / 3)  # Kết quả: 3.3333333333333335
print(10 // 3) # Kết quả: 3
```

### **2.2. Modulo (`%`)**

- Toán tử này không trả về thương số của phép chia, mà trả về phần dư còn lại sau khi chia.
- Nó thường được gọi là **remainder operator**.

```python
total_cookies = 10
friends = 3
leftover_cookies = total_cookies % friends
print(leftover_cookies) # Kết quả: 1 (Sau khi chia mỗi người 3 cái, còn dư 1 cái)

number = 17
divisor = 5
remainder = number % divisor
print(remainder) # Kết quả: 2 (17 chia 5 được 3 và dư 2)
```

- **Toán tử modulo** thường được dùng để kiểm tra một số là chẵn hay lẻ. Một số là chẵn nếu `number % 2 == 0`, và là lẻ trong các trường hợp còn lại.

```python
num = 6
print(num % 2) # Kết quả: 0 (Vì 6 là số chẵn)

num = 7
print(num % 2) # Kết quả: 1 (Vì 7 là số lẻ)
```

## **3. Exponentiation**

- Để nâng một số lên lũy thừa của một số khác, bạn dùng toán tử hai dấu sao (`**`). VD: $x^y$ sẽ được viết `x**y`

```python
base = 2
exponent = 3
power_result = base ** exponent # Tương đương 2 * 2 * 2
print(power_result) # Kết quả: 8

side = 5
area_square = side ** 2 # Tương đương 5 * 5 (bình phương)
print(area_square) # Kết quả: 25
```

## **4. Order of Operations**

- Khi một biểu thức chứa nhiều toán tử, Python tuân theo **thứ tự thực hiện phép toán (order of operations)** chuẩn, tương tự toán học.
- Thứ tự này thường được ghi nhớ bằng các chữ viết tắt như **PEMDAS/BODMAS**:

  1. Parentheses `()`
  2. Exponentiation `**`
  3. Multiplication `*`, Division `/`, Floor Division `//`, Modulo `%`
  4. Addition `+`, Subtraction `-`

```python
result = 2 + 3 * 4
print(result) # Kết quả: 14 (Nhân được thực hiện trước cộng: 2 + 12)
```

- Nếu bạn muốn phép cộng diễn ra trước, hãy dùng ngoặc:

```python
result = 2 + 3 * 4
print(result) # Kết quả: 14 (Nhân được thực hiện trước cộng: 2 + 12)
```

- Một ví dụ khác:

```python
calculation = 100 - 20 / 5 * 2 + 3**2
# Đánh giá từng bước:
# 1. Lũy thừa: 3**2 trở thành 9
#    Biểu thức: 100 - 20 / 5 * 2 + 9
# 2. Chia/Nhân (từ trái sang phải):
#    20 / 5 trở thành 4.0
#    Biểu thức: 100 - 4.0 * 2 + 9
#    4.0 * 2 trở thành 8.0
#    Biểu thức: 100 - 8.0 + 9
# 3. Cộng/Trừ (từ trái sang phải):
#    100 - 8.0 trở thành 92.0
#    Biểu thức: 92.0 + 9
#    92.0 + 9 trở thành 101.0
print(calculation) # Kết quả: 101.0
```

- Sử dụng ngoặc `()` có thể làm cho biểu thức phức tạp trở nên rõ ràng hơn, ngay cả khi theo quy tắc ưu tiên thì chúng không bắt buộc.
- Khi phân vân, hãy dùng ngoặc để đảm bảo phép tính diễn ra theo đúng thứ tự bạn mong muốn.

## **5. Shorthand: Augmented Assignment Operators**

- Trong lập trình, việc cập nhật một biến dựa trên giá trị hiện tại của nó là rất phổ biến. Ví dụ, tăng một bộ đếm hoặc cộng dồn một tổng:

```python
count = 0
count = count + 1 # Tăng count lên 1
print(count) # Kết quả: 1

total_cost = 50
item_price = 15.50
total_cost = total_cost + item_price # Cộng item_price vào total_cost
print(total_cost) # Kết quả: 65.5
```

- Python cung cấp các toán tử viết tắt, gọi là **toán tử gán kết hợp (augmented assignment operators)**, để diễn đạt ngắn gọn hơn:

```python
count = 0
count += 1 # Tương đương count = count + 1
print(count) # Kết quả: 1

score = 100
score -= 10 # Tương đương score = score - 10
print(score) # Kết quả: 90

multiplier = 5
multiplier *= 2 # Tương đương multiplier = multiplier * 2
print(multiplier) # Kết quả: 10

total = 50.0
total /= 4 # Tương đương total = total / 4
print(total) # Kết quả: 12.5
```

- Các toán tử viết tắt này (`+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`) tồn tại cho tất cả các toán tử số học và kết hợp phép toán với phép gán. Chúng rất tiện lợi và thường giúp mã dễ đọc hơn khi bạn đã quen với chúng.
