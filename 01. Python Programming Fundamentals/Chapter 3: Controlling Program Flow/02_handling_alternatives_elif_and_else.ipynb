# **Handling Alternatives: elif and else**

- **Câu lệnh `if` (`if` statement)** cho phép chương trình thực thi một **khối mã (code block)** chỉ khi một **điều kiện (condition)** cụ thể được đánh giá là `True`.
- uy nhiên, trong thực tế, chương trình thường cần kiểm tra nhiều điều kiện có thể xảy ra hoặc thực hiện một **hành động mặc định (default action)** khi điều kiện ban đầu không được thỏa mãn. Các mệnh đề `elif` và `else` mở rộng câu lệnh `if` để xử lý những tình huống ra **quyết định (decision-making)** phức tạp hơn này.

## **1. Checking Multiple Conditions with `elif`**

- Hãy tưởng tượng bạn muốn kiểm tra nhiều hơn một điều kiện. Ví dụ, để xác định một số là dương, âm hay bằng 0 thì cần kiểm tra nhiều khả năng khác nhau. Câu lệnh `elif`, viết tắt của **“else if”**, cho phép bạn kiểm tra các **điều kiện bổ sung (additional conditions)** chỉ khi các điều kiện `if` (hoặc `elif`) đứng trước đó đều là `False`.

- Bạn có thể nối nhiều câu lệnh `elif` sau một câu lệnh `if` ban đầu. Python đánh giá **các điều kiện (conditions)** lần lượt từ trên xuống dưới. Ngay khi tìm thấy một điều kiện là `True`, nó sẽ thực thi **khối mã (code block)** tương ứng với điều kiện đó rồi bỏ qua phần còn lại của toàn bộ cấu trúc `if-elif-else`.

- Cấu trúc tổng quát:
    ```python
    if condition_1:
        # Khối 1: Thực thi nếu condition_1 là True
        statement(s)
    elif condition_2:
        # Khối 2: Thực thi nếu condition_1 là False VÀ condition_2 là True
        statement(s)
    elif condition_3:
        # Khối 3: Thực thi nếu condition_1 VÀ condition_2 đều False,
        # VÀ condition_3 là True
        statement(s)
    # ... bạn có thể có thêm nhiều khối elif khác
    ```

````python
number = -10

if number > 0:
    print("The number is positive.")
elif number < 0:
    print("The number is negative.")

# Nếu number là 0, thì cả hai điều kiện trên đều không True,
# nên hiện tại sẽ không có gì xảy ra.
````

- Trong trường hợp này, nếu `number = -10`:
  
  - `number > 0` là `False`, nên khối đầu tiên bị bỏ qua.
  - `number < 0` là `True`, nên thông báo `"The number is negative."` được in ra. Sau đó chương trình bỏ qua mọi mệnh đề `elif` hoặc `else` tiếp theo trong cấu trúc này.

  - Nếu số là `0` thì cả `number > 0` lẫn `number < 0 đều` đúng là `False`. Trong đoạn mã trên, sẽ không có gì được in ra. Để xử lý trường hợp không có điều kiện nào trong `if` hoặc `elif` được thỏa mãn, ta dùng mệnh đề `else` làm mặc định.

## **2. The Default Case: `else`**

- Mệnh đề `else` cung cấp một **khối mã mặc định (default block)** chỉ được thực thi khi tất cả các điều kiện `if` và `elif` đứng trước đó đều đánh giá là `False`.

- Nó đóng vai trò như một **nhánh bao quát (catch-all)**. Nếu được sử dụng, `else` phải luôn đứng cuối cùng, sau tất cả các mệnh đề `if` và `elif`, và nó không đi kèm với điều kiện nào.

- Cấu trúc hoàn chỉnh:
    ```python
    if condition_1:
        # Khối 1: Thực thi nếu condition_1 là True
        statement(s)
    elif condition_2:
        # Khối 2: Thực thi nếu condition_1 là False VÀ condition_2 là True
        statement(s)
    # ... thêm các khối elif khác (không bắt buộc) ...
    else:
        # Khối N: Thực thi nếu TẤT CẢ các điều kiện phía trước đều False
        statement(s)
    ```

````python
number = 0

print(f"Checking the number: {number}")

if number > 0:
    print("The number is positive.")
elif number < 0:
    print("The number is negative.")
else:
    # Khối này chỉ chạy nếu number không > 0 VÀ cũng không < 0
    print("The number is zero.")

print("Finished checking.")
````

- Nếu chạy với `number = 0`:

  - `number > 0` là `False`.
  - `number < 0` là `False`.

- Vì tất cả các điều kiện trước đó đều `False`, khối else được thực thi và in ra `"The number is zero."`

## **3. Order and Exclusivity**

- Điều quan trọng cần nhớ về cấu trúc `if-elif-else` là:

### **3.1. Order Matters**

- Các điều kiện được kiểm tra theo đúng thứ tự xuất hiện. Nếu có nhiều điều kiện có thể `True`, chỉ khối gắn với điều kiện `True` đầu tiên được thực thi. Ví dụ, xét trường hợp xếp loại sau:

````python
score = 85

if score >= 70: # Điều kiện này đúng trước tiên
    grade = 'C'
elif score >= 80: # Điều kiện này cũng đúng, nhưng được kiểm tra sau
    grade = 'B'
else:
    grade = 'Not C or B'

print(grade) # Output: C
````

- Ngay khi điều này xảy ra: `score >= 70` được kiểm tra trước và đúng với `85` thì biến `grade` được gán là `'C'`:

  - Python thực thi khối tương ứng
  - toàn bộ các `elif` phía sau bị bỏ qua

- Để có kết quả xếp loại đúng, bạn nên sắp xếp các điều kiện từ **cụ thể nhất (most specific)** đến **ít cụ thể hơn (less specific)**.

````python
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80: # Được kiểm tra thứ hai, đúng với 85
    grade = 'B'
elif score >= 70: # Được kiểm tra thứ ba
    grade = 'C'
# ... các mức khác ...
else:
    grade = 'F'

print(grade) # Output: B
````

### **3.2 Only One Block Executes**

-  Trong một chuỗi `if-elif-else`, tối đa chỉ có một block được thực thi. Dù các điều kiện `elif` phía sau cũng có thể đúng, chúng vẫn bị bỏ qua ngay khi một điều kiện `if` hoặc `elif` trước đó đã đúng. Khối `else` chỉ chạy khi không có điều kiện nào phía trên đúng.
