# **Making Decisions: The if Statement**

- Trong lập trình Python, khi bạn muốn một đoạn code chỉ **thực hiện logic** khi nó **thỏa mãn một điều kiện nào đó**.
- Ví dụ: Ở ngoài đời thực - Nếu trời mưa, tôi sẽ mang ô. Chương trình bạn viết code cũng cần cách để ra quyết định tương tự. Và Python cung cấp **câu lệnh `if`** để làm điều này.

- **Câu lệnh `if`** cho phép bạn thực thi một **khối mã (code block)** có **điều kiện**. Nó kiểm tra xem một điều kiện nào đó có `True` hay `False`.
    - Nếu **điều kiện** là `True` -> **code block** của `if` sẽ được thực thi.
    - Nếu **điều kiện** là `False` -> **code block** bị bỏ qua, và chương trình tiếp tục các lệnh tiếp theo.

## **1. The Structure of an `if` Statement**

- Cú pháp cơ bản như sau:

```python
if condition:
    # Mã được thực thi nếu điều kiện là True
    statement1
    statement2
    # ... nhiều lệnh khác ...

# Mã ở đây sẽ chạy bất kể điều kiện đúng hay sai, sau khối if
next_statement
```

- Ta hãy phân tích:
    1. **`if` Keyword**: Câu lệnh bắt đầu bằng từ khóa `if`.
    2. **`condition`**: Đây là một biểu thức trả về `True` hoặc `False`. Nó thường được gọi là **boolean expression** và dùng:
        - Các **comparison operators** như `==`,`!=`,`<`,`>`,`<=`,`>=`
        - Hoặc các **logical operators** như `and`, `or`, `not`
    3. **Colon `:`**: Sau điều kiện phải có dấu hai chấm. Nó đánh dấu bắt đầu của **(code block)** phụ thuộc vào điều kiện.
    4. **Indented Block**:
        - Các dòng code ngay sau `if condition`: và được thụt vào bên phải sẽ tạo thành **if block**.
        - Python dùng **indentation (thường là 4 dấu cách)** để xác định các **(code block)**. Tất cả các dòng có cùng mức thụt lề dưới `if` đều thuộc cùng một block.

## **2. How it Works**

- Hãy tưởng tượng bạn đang viết một chương trình kiểm tra xem người dùng đã đủ tuổi để bỏ phiếu chưa. Độ tuổi bỏ phiếu thường là 18

````python
age = 20

print("Checking voting eligibility...")

if age >= 18:
    print("You are old enough to vote.")
    print("Please register if you haven't already.")

print("Eligibility check complete.")
````

- Biến `age` được gán giá trị `20`.
- Câu lệnh `print("Checking voting eligibility...")` được thực thi.
- **Câu lệnh `if age >= 18:`** kiểm tra điều kiện. Vì `20 >= 18` là `True`, **điều kiện được thỏa mãn**.
- Do **điều kiện đúng**, **code block bên trong `if`** được thực thi:
    - `print("You are old enough to vote.")`
    - `print("Please register if you haven't already.")`
- Sau khi **khối`if`** kết thúc, chương trình tiếp tục thực thi **`print("Eligibility check complete.")`**.

- Nhưng trong trường hợp tuổi người dùng dưới 18 thì sẽ như nào?

````python
age = 15

print("Checking voting eligibility...")

if age >= 18:
    # Khối này sẽ bị bỏ qua
    print("You are old enough to vote.")
    print("Please register if you haven't already.")

print("Eligibility check complete.")
````

- Biến `age` được gán giá trị `15`.
- Câu lệnh `print("Checking voting eligibility...")` được thực thi.
- **Câu lệnh `if age >= 18:`** kiểm tra điều kiện. Vì `15 >= 18` là `False`, **điều kiện không được thỏa mãn**.
- Do **điều kiện sai**, **code block bên trong `if`** bị bỏ qua.
- Chương trình tiếp tục thực thi **`print("Eligibility check complete.")`**.

## **3. The Importance of Indentation**
- Indentation trong Python không chỉ để dễ đọc; nó là một phần của cú pháp. Nó cho interpreter biết câu lệnh nào thuộc về khối if. Thụt lề sai sẽ dẫn đến lỗi hoặc hành vi không mong muốn.

````python
# Thụt lề đúng
temperature = 30
if temperature > 25:
    print("It's a warm day.")   # Thụt lề đúng
    print("Wear light clothes.") # Thụt lề đúng

# Thụt lề sai (sẽ gây ra IndentationError)
if temperature > 25:
print("It's a warm day.") # Thiếu thụt lề
````

- Câu lệnh `if` là cách đơn giản nhất để đưa logic có điều kiện vào chương trình của bạn. Nó cho phép mã phản ứng khác nhau dựa trên các điều kiện cụ thể, khiến chương trình trở nên linh hoạt và “thông minh” hơn. Như bạn sẽ thấy tiếp theo, bạn có thể mở rộng nền tảng này bằng `elif` và `else` để xử lý các tình huống thay thế.
