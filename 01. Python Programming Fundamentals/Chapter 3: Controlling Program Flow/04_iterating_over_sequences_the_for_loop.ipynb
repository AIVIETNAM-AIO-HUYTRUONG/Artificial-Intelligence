# **Iterating Over Sequences: The for Loop**

- Python cung cấp nhiều cấu trúc để kiểm soát **luồng thực thi chương trình (program flow)**. **Vòng lặp for (for loop)** cung cấp một cách trực tiếp và hiệu quả để thực hiện một hành động cho từng phần tử trong một **dãy (sequence)** hoặc **tập hợp (collection)**

- Nó đặc biệt hữu ích khi xử lý một collection các phần tử đã biết trước, từng phần tử một. Những trường hợp dùng phổ biến bao gồm lặp qua tất cả các phần tử trong **danh sách (list)**, hoặc tất cả ký tự trong một **string**, hoặc lặp lại một hành động cụ thể.

## **1. Basic `for` Loop Syntax**

- Cấu trúc nền tảng của vòng lặp `for` như sau:
  ```python
  for variable_name in sequence:
      # Khối mã sẽ được thực thi cho mỗi phần tử
      # Dùng variable_name để truy cập phần tử hiện tại
      statement1
      statement2
      # ...
  ```

- Hãy phân tích:

- `for`: Từ khóa bắt đầu vòng lặp.
- `variable_name`: Một biến do bạn chọn (nên đặt tên có ý nghĩa) để chứa phần tử hiện tại từ dãy trong mỗi lần lặp (iteration).
- `in`: Từ khóa nối giữa biến vòng lặp và dãy.
- `sequence`: Tập hợp các phần tử mà bạn muốn lặp qua. Nó có thể là list, strings, hoặc các kiểu iterable khác mà ta sẽ gặp sau.
- **Colon**`:` : Dấu hai chấm kết thúc dòng lệnh `for`.
- **Khối mã thụt lề (indented code block)**: Tương tự như `if` và `while`, phần mã nằm trong vòng lặp phải được thụt lề. Khối này sẽ được thực thi một lần cho mỗi phần tử trong `sequence`.

## **2. Iterating Over Lists**

- Lists là một **cấu trúc dữ liệu có thứ tự (ordered collection)**, nên rất phù hợp với vòng lặp `for`. Hãy tưởng tượng bạn có một danh sách nhiệt độ và muốn in từng giá trị.

````python
temperatures = [19.5, 22.1, 18.0, 25.3]

print("Daily temperatures:")
for temp in temperatures:
    print(f"Temperature recorded: {temp}°C")

print("\nLoop finished.")
````

- Trong ví dụ này:

  - `temperatures` là **dãy (sequence)** — một danh sách.
  - `temp` là **biến vòng lặp (loop variable)**.
  - Ở lần lặp đầu tiên, `temp` nhận giá trị `19.5`. Câu lệnh print thụt lề được thực  thi.
  - Ở lần lặp thứ hai, `temp` nhận giá trị `22.1`. Câu lệnh print lại được thực thi.
  - Quá trình này tiếp tục cho đến khi phần tử cuối cùng (`25.3`) được xử lý.
  - Sau khi vòng lặp kết thúc, chương trình tiếp tục với dòng không thụt lề tiếp  theo (`print("\nLoop finished.")`).

## **3. Iterating Over Strings**

- Chuỗi là một dãy các ký tự. Vòng lặp `for` có thể tự động lặp qua từng ký tự trong một chuỗi.

````python
user_name = "Alice"

print(f"Characters in the name {user_name}:")
for character in user_name:
    print(f"- {character}")
````

- Ở đây, biến vòng lặp `character` lần lượt nhận giá trị của từng ký tự (`'A'`, rồi `'l'`, rồi `'i'`, v.v.) trong chuỗi `user_name` ở mỗi lần lặp.

## **4. Repeating Code a Specific Number of Times with `range()`**

- Đôi khi, bạn không có sẵn lists hay strings để lặp qua, nhưng chỉ muốn lặp lại một hành động một số lần cố định. **Hàm dựng sẵn (built-in function)** `range()` của Python rất hữu ích trong trường hợp này. Nó tạo ra một dãy số mà vòng lặp for có thể lặp qua.

### **4.1. `range(stop)`**

- `range(stop)`: Tạo các số bắt đầu từ `0` đến trước giá trị `stop`.

````python
print("Counting from 0 to 4:")
for i in range(5): # Generates 0, 1, 2, 3, 4
    print(i)
````

- **Lưu ý**:
  -  `range(5)` tạo ra các số từ `0` đến `4`, tức là tổng cộng 5 con số. Cách đánh số bắt đầu từ 0 này rất phổ biến trong lập trình.
  - Biến `i` thường được dùng làm **loop counter** đơn giản, nhưng bạn có thể dùng bất kỳ tên biến hợp lệ nào.

### **4.2 `range(start, stop)`**

- `range(start, stop)`: Tạo các số bắt đầu từ `start` đến trước giá trị `stop`.

````python
print("\nNumbers from 2 to 5:")
for num in range(2, 6): # Generates 2, 3, 4, 5
    print(num)
````

### **4.3 `range(start, stop, step)`**

- Tạo ra các số bắt đầu từ `start`, đến trước `stop`, tăng thêm `step` sau mỗi lần.

````python
print("\nOdd numbers less than 10:")
for odd_num in range(1, 10, 2): # Generates 1, 3, 5, 7, 9
    print(odd_num)
````

- Dùng `range()` với `for` loop là cách chuẩn để thực hiện các việc như chạy mô phỏng trong 100 bước, xử lý 10 phần tử đầu tiên của một dataset, hoặc đơn giản là in một thông báo nhiều lần.

## **5. Practical Example: Summing List Elements**
- Hãy kết hợp vòng lặp `for` với những gì ta đã học về biến và toán tử để tính tổng các số trong một danh sách.

````python
scores = [88, 92, 75, 98, 85]
total_score = 0 # Khởi tạo biến tích lũy (*accumulator variable*)

for score in scores:
    total_score = total_score + score # Cộng điểm hiện tại vào tổng

print(f"The list of scores is: {scores}")
print(f"The total score is: {total_score}")
````

- Trong ví dụ này, `total_score` đóng vai trò là một **biến tích lũy (accumulator)**. Nó bắt đầu từ `0`, và ở mỗi lần lặp, điểm hiện tại từ danh sách `scores` được cộng vào nó.

## **6. When to Use `for` vs `while`**

- Dùng **`for` loop**:
  - Khi bạn muốn lặp qua mọi item trong một **sequence** đã biết trước (như `list`, `string`, hoặc `range()`)
  - Hoặc khi bạn cần lặp một hành động một số lần xác định. for sẽ tự quản lý quá trình lặp cho bạn.

- Dùng **`while` loop**:
  - Khi bạn cần lặp một hành động chừng nào một **condition** còn đúng, và bạn không nhất thiết biết trước loop sẽ chạy bao nhiêu lần

- Vòng lặp `for` cung cấp một cách ngắn gọn và dễ đọc để xử lý các phần tử trong dãy, khiến nó trở thành một trong những **cấu trúc điều khiển luồng (control flow structures)** được dùng nhiều nhất trong Python.
- Khi bạn học các cấu trúc dữ liệu phức tạp hơn như **từ điển (dictionary)** và **tuple** ở chương tiếp theo, bạn sẽ thấy rằng vòng lặp `for` cũng hoạt động rất tốt với chúng.
