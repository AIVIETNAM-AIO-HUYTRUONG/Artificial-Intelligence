# **Controlling Loops: break and continue**

- Mặc dù vòng lặp `for` và `while` cung cấp những cách mạnh mẽ để lặp lại các **khối mã (code block)**, đôi khi bạn cần khả năng kiểm soát chi tiết hơn đối với quá trình thực thi của chúng:
  
  - Bạn có thể muốn thoát khỏi vòng lặp ngay lập tức trước khi nó hoàn thành
  - Bạn có thể bỏ qua phần còn lại của lần lặp hiện tại và chuyển sang lần lặp tiếp theo.

- Python cung cấp hai câu lệnh dành riêng cho những tình huống này: `break` và `continue`.

## **1. Exiting Early: The `break` Statement**

- Câu lệnh `break` dùng để kết thúc vòng lặp hiện tại một cách sớm (prematurely terminate the loop). Ngay khi Python gặp `break` bên trong vòng lặp `for` hoặc `while`, nó sẽ thoát khỏi vòng lặp ngay lập tức.

- Điều này đặc biệt hữu ích khi bạn đang tìm kiếm một giá trị và muốn dừng lại ngay khi tìm thấy, hoặc khi xuất hiện một điều kiện lỗi khiến việc tiếp tục lặp trở nên vô nghĩa.

- Ví dụ: tìm lần xuất hiện đầu tiên của một số trong danh sách.

```python
numbers = [1, 5, 12, 8, 9, 21, 5, 3]
target = 9
found_at_index = -1 # Khởi tạo với giá trị biểu thị "không tìm thấy"

print(f"Searching for {target} in {numbers}...")

for index, num in enumerate(numbers):
    print(f"Checking index {index}: value {num}")

    if num == target:
        found_at_index = index
        print(f"Found {target}!")
        break # Thoát vòng lặp ngay lập tức

# Mã ở đây chạy sau khi vòng lặp kết thúc hoặc bị break
if found_at_index != -1:
    print(f"Target {target} first found at index {found_at_index}.")
else:
    print(f"Target {target} not found in the list.")

print("Search complete.")
```

- Hãy chú ý rằng vòng lặp đã dừng ở **chỉ mục (index)**: `4`. Các giá trị `21`, `5`, và `3` không hề được kiểm tra vì câu lệnh `break` đã khiến vòng lặp thoát ngay khi điều kiện `num == target` trở thành `True`.

## **2. Skipping an Iteration: The `continue` Statement**

- Câu lệnh `continue` hoạt động khác với `break`. Thay vì kết thúc toàn bộ vòng lặp, `continue` sẽ bỏ qua phần còn lại của mã trong **lần lặp hiện tại (current iteration)** và ngay lập tức chuyển sang lần lặp tiếp theo.
  
  - Với vòng lặp `while`, điều này có nghĩa là quay lại kiểm tra điều kiện của vòng lặp.
  - Với vòng lặp `for`, điều này có nghĩa là chuyển sang phần tử tiếp theo trong dãy.

- Điều này hữu ích khi bạn gặp dữ liệu hoặc điều kiện mà bạn muốn bỏ qua mà không cần dừng xử lý các phần tử còn lại.

- Ví dụ: tính tổng chỉ các số dương trong một danh sách.

```python
data = [10, -5, 20, 0, -15, 30, 5]
positive_sum = 0

print(f"Processing data: {data}")

for num in data:
    print(f"Current number: {num}")

    if num <= 0:
        print("  Skipping non-positive number.")
        continue # Bỏ qua phần còn lại của lần lặp này

    # Đoạn mã này chỉ chạy nếu continue KHÔNG được thực thi
    print(f"  Adding {num} to sum.")
    positive_sum += num

print(f"\nSum of positive numbers: {positive_sum}")
```

- Ở đây, mỗi khi điều kiện `num <= 0` là `True`, câu lệnh `continue` được thực thi. Điều này ngăn dòng `positive_sum += num` chạy với các giá trị `-5`, `0`, và `-15`, nhưng vòng lặp vẫn tiếp tục xử lý các phần tử còn lại (`30` và `5`).

## **3. `break` vs `continue`**

- `break` : Thoát hoàn toàn khỏi vòng lặp. Chương trình tiếp tục chạy sau vòng lặp
- `continue` : Bỏ qua lần lặp hiện tại. Chương trình tiếp tục với lần lặp kế tiếp.

## **4. Nested Loops**

- Điều quan trọng cần nhớ là break và continue chỉ ảnh hưởng đến **vòng lặp gần nhất (innermost loop)** chứa chúng.

- Nếu bạn có vòng lặp lồng nhau (một vòng lặp nằm bên con vòng lặp cha), thì `break` hoặc `continue` trong vòng lặp bên con sẽ không ảnh hưởng đến vòng lặp cha bên ngoài.

```python
for i in range(1, 4): # Vòng lặp cha
    print(f"Outer loop iteration {i}:")

    for j in ['a', 'b', 'c', 'STOP', 'd']: # Vòng lặp con

        if j == 'STOP':
            print("  Inner loop breaking!")
            break # Chỉ thoát vòng lặp con

        if j == 'b':
            print(f"  Skipping '{j}' in inner loop...")
            continue # Chuyển sang phần tử tiếp theo con vòng lặp con

        print(f"  Processing inner item: {j}")

    # Dòng này chạy sau khi vòng lặp con kết thúc hoặc bị break
    print(f"Outer loop iteration {i} finished.")

print("All loops complete.")
```

- Việc thành thạo `break` và `continue` giúp bạn kiểm soát chính xác cách các vòng lặp hoạt động, cho phép xử lý hiệu quả các điều kiện đặc biệt và làm cho chương trình phản ứng linh hoạt hơn với dữ liệu.
