# **Nested Control Structures**

- Chương trình có thể thực thi mã có điều kiện bằng `if`, `elif`, `else`, và lặp lại hành động bằng các vòng lặp `while` và `for`. Khi kết hợp các cấu trúc này với nhau, chúng ta có thể xây dựng những logic mạnh mẽ và phức tạp hơn. Các câu **lệnh điều khiển luồng (control flow statements)** có thể được đặt bên trong các câu lệnh điều khiển luồng khác. Kỹ thuật này được gọi là **lồng nhau (nesting)**.

- Hãy tưởng tượng bạn cần kiểm tra nhiều điều kiện trước khi thực hiện một hành động, hoặc cần thực hiện một tác vụ lặp đi lặp lại cho từng phần tử trong một vòng lặp khác. Cấu trúc lồng nhau giúp thực hiện điều đó.

## **1. Nested Conditional Statements**

- Bạn có thể đặt một câu lệnh `if`, `elif`, hoặc `else` bên trong một **khối `if`, `elif`, hoặc `else`** khác.
- Điều này cho phép kiểm tra các **điều kiện phụ (secondary conditions)** chỉ khi **điều kiện chính (primary condition)** được thỏa mãn.

- Ví dụ: kiểm tra xem một người có đủ điều kiện nhận giảm giá hay không — họ phải là sinh viên và có thẻ hợp lệ.

```python
is_student = True
has_valid_id = False
age = 20

print("Checking discount eligibility...")

if age < 25: # Điều kiện ngoài: kiểm tra tuổi
    print("Age requirement met.")

    if is_student: # Điều kiện trong 1: kiểm tra sinh viên
        print("Student status confirmed.")

        if has_valid_id: # Điều kiện trong 2: kiểm tra thẻ
            print("Valid ID confirmed.")
            print("Congratulations! You are eligible for the discount.")

        else: # else bên trong (thuộc kiểm tra ID)
            print("Valid ID not found. Discount not applicable.")

    else: # else bên trong (thuộc kiểm tra sinh viên)
        print("Not a student. Discount not applicable.")

else: # else bên ngoài (thuộc kiểm tra tuổi)
    print("Age requirement not met. Discount not applicable.")

print("Eligibility check complete.")
```

- Hãy chú ý đến **thụt lề (indentation)**. Các  `if/else` **blocks** bên trong được thụt vào sâu hơn so với khối bên ngoài. Python dùng thụt lề để xác định câu lệnh nào thuộc về block nào.

- Các kiểm tra bên trong (tình trạng sinh viên, thẻ hợp lệ) chỉ được thực hiện nếu điều kiện bên ngoài `age < 25` là `True`.

## **2. Nested Loops**

- Các vòng lặp cũng có thể được lồng vào nhau. Một trường hợp sử dụng phổ biến là duyệt qua các **cấu trúc dữ liệu nhiều chiều (multi-dimensional data structures)** như bảng hoặc lưới.

### **2.1. Nested `for` Loops**

- Giả sử bạn muốn in tất cả các cặp tọa độ `(x, y)` trong một lưới 3x2.

```python
print("Generating grid coordinates:")

# Vòng lặp ngoài duyệt giá trị x
for x in range(3): # x = 0, 1, 2

    # Vòng lặp trong duyệt giá trị y cho MỖI x
    for y in range(2): # y = 0, 1
        print(f"  ({x}, {y})")

print("Coordinate generation finished.")
```

- Cách thực thi:

  1. Vòng lặp cha bắt đầu với `x = 0`.
  2. Vòng lặp con chạy hoàn toàn với `x = 0`. Lần lượt:
    - `y = 0` → in `(0, 0)`
    - `y = 1` → in `(0, 1)`
  3. Vòng lặp con kết thúc `x = 0`.
  4. Vòng lặp cha chuyển sang lần lặp tiếp theo, đặt `x = 1`.
  5. Vòng lặp con lại chạy hoàn toàn với `x = 1`. Lần lượt
    - `y = 0` -> in `(1,0)`
    - `y = 1` -> in `(1,1)`
  6. Vòng lặp con kết thúc `x = 1`.
  7. Vòng lặp cha lại tiếp tục `x = 2`.
  8. Vòng lặp con lại chạy hoàn toàn với `x = 2`. Lần lượt
    - `y = 0` -> in `(2,0)`
    - `y = 1` -> in `(2,1)`
  9. Kết thúc 2 vòng lặp.

- Điều quan trọng cần nhớ:

  > **Vòng lặp bên con (inner loop)** sẽ hoàn thành toàn bộ các lần lặp của nó cho mỗi một lần lặp của **vòng lặp bên cha (outer loop)**.

### **2.2. Nested while Loops**

- Bạn cũng có thể lồng các vòng lặp `while`, mặc dù cách này đòi hỏi phải quản lý cẩn thận các điều kiện vòng lặp để tránh **vòng lặp vô hạn (infinite loop)**.

```python
outer_count = 0
print("Outer loop starting...")

while outer_count < 2:
    print(f"Outer loop iteration {outer_count}")
    inner_count = 0

    # Vòng lặp trong chạy khi inner_count < 3
    while inner_count < 3:
        print(f"  Inner loop iteration {inner_count}")
        inner_count += 1 # Quan trọng: cập nhật biến của vòng lặp trong

    outer_count += 1 # Quan trọng: cập nhật biến của vòng lặp ngoài

print("Outer loop finished.")
```

- Cũng giống như vòng lặp `for` lồng nhau, vòng lặp `while` bên trong sẽ chạy đến khi hoàn tất cho mỗi lần lặp của vòng lặp bên ngoài.

- Hãy đảm bảo rằng điều kiện của cả hai vòng lặp cuối cùng đều sẽ trở thành `False`.

## **3. Combining Loops and Conditionals**

- Một trong những mẫu phổ biến nhất là lồng các câu lệnh điều kiện bên trong vòng lặp, hoặc lồng vòng lặp bên trong câu lệnh điều kiện.

### **3.1. `if` Statements Inside Loops**

- Đây là mẫu rất thường gặp. Bạn duyệt qua một dãy và chỉ thực hiện hành động khi phần tử đó thỏa mãn một điều kiện nào đó.

- Ví dụ: in ra chỉ các số chẵn trong một danh sách.

```python
numbers = [11, 22, 37, 44, 58, 61, 76]
print("Even numbers found:")

for num in numbers:
    # Kiểm tra điều kiện cho từng số trong danh sách
    if num % 2 == 0: # Số đó có chia hết cho 2 không?
        print(f"  {num}")

print("Search complete.")
```

- Vòng lặp `for` duyệt qua từng `num` trong danh sách `numbers`. Bên trong vòng lặp, câu lệnh `if` kiểm tra xem số hiện tại có chẵn không. Chỉ khi điều kiện `num % 2 == 0` là `True` thì lệnh `print()` mới được thực thi.

### **3.2. Loops Inside `if` Statements**

- Đôi khi bạn muốn chỉ thực thi vòng lặp khi một điều kiện nhất định là đúng.

-  Ví dụ: nếu một danh sách không rỗng, hãy in nội dung của nó.

```python
data = ["apple", "banana", "cherry"]
# data = [] # Hãy thử bỏ chú thích dòng này để thấy sự khác biệt

if data: # Kiểm tra danh sách không rỗng (danh sách không rỗng là truthy)
    print("Processing data items:")
    for item in data:
        print(f"  - {item}")
else:
    print("No data items to process.")
```

- Ở đây, vòng lặp `for` được lồng bên trong khối `if`. Nó chỉ chạy nếu kiểm tra `if data:` ban đầu thành công (nghĩa là danh sách không rỗng).

## **4. The Importance of Indentation**

- Hãy nhớ rằng Python sử dụng thụt lề để xác định các khối mã. Với các cấu trúc lồng nhau, thụt lề chính xác là điều tuyệt đối cần thiết. Mỗi cấp độ lồng nhau cần thêm một mức thụt lề (thường là 4 dấu cách).

- Thụt lề sai sẽ dẫn đến lỗi `IndentationError`, hoặc tệ hơn là mã chạy nhưng cho ra kết quả sai vì logic không được Python hiểu như bạn mong muốn.

```python
# Thụt lề đúng
for i in range(2):
    print(f"Outer {i}")
    if i == 0:
        print("  Inner check for i=0") # Thụt lề dưới if
    for j in range(2):
        print(f"    Inner loop {j}") # Thụt lề dưới vòng lặp trong

# Thụt lề sai (rất có thể gây lỗi hoặc logic sai)
for i in range(2):
print(f"Outer {i}") # Lỗi: Expected an indented block
if i == 0:
print("  Inner check for i=0")
for j in range(2):
print(f"    Inner loop {j}") # Lỗi hoặc phạm vi sai
```

## **5. Keeping Nesting Readable**

- Mặc dù lồng nhau rất mạnh mẽ, các cấu trúc lồng quá sâu:
  - ví dụ: một vòng lặp bên trong một vòng lặp bên trong một `if` bên trong một vòng lặp khác có thể trở nên rất khó đọc và khó hiểu.

- Thông thường, bạn nên cố gắng giới hạn mức lồng vào khoảng hai hoặc ba cấp. Nếu bạn thấy mình cần lồng sâu hơn, hãy tách logic thành những hàm nhỏ hơn, riêng biệt.
