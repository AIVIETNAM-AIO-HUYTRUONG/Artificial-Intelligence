# **Accessing and Modifying Dictionary Data**

- Dictionary lưu trữ thông tin bằng cách sử dụng các **key duy nhất (unique keys)** được liên kết với các value. Cấu trúc này rất hiệu quả khi cần tra cứu thông tin và đã biết key. Các phương pháp truy xuất, thay đổi, thêm và xóa các cặp **key-value** được trình bày dưới đây..

## **1. Accessing Values Using Keys**

- Cách trực tiếp nhất để lấy value được liên kết với một key cụ thể là sử dụng dấu ngoặc vuông `[]`. Cách này tương tự như việc truy cập phần tử trong list bằng index, nhưng ở đây bạn sử dụng key thay vì index.

```python
# Example dictionary: Student grades
student_grades = {'Alice': 85, 'Bob': 92, 'Charlie': 78}

# Access Bob's grade
bob_grade = student_grades['Bob']
print(f"Bob's grade is: {bob_grade}") # Output: Bob's grade is: 92

# Access Alice's grade
alice_grade = student_grades['Alice']
print(f"Alice's grade is: {alice_grade}") # Output: Alice's grade is: 85
```

- Cần cẩn thận: nếu bạn cố gắng truy cập một key không tồn tại trong Dictionary bằng dấu ngoặc vuông, Python sẽ phát sinh `KeyError`. Lỗi này cho biết key mà bạn yêu cầu không được tìm thấy.

```python
# Trying to access a non-existent key
david_grade = student_grades['David'] # This line would cause a KeyError
print(david_grade)
```

- Trong Chương 9, bạn sẽ học cách xử lý những lỗi như vậy một cách phù hợp. Tuy nhiên, ở thời điểm hiện tại, bạn chỉ cần biết rằng lỗi này có thể xảy ra.

## **2. Safer Ways to Access Data**

- Do việc truy cập một key không tồn tại bằng `[]` sẽ gây lỗi, Python cung cấp các cách an toàn hơn để truy xuất value hoặc kiểm tra sự tồn tại của key.

### **2.1. Using the `get()` Method**

- Phương thức `get()` của Dictionary truy xuất value của một key tương tự như dấu ngoặc vuông. Tuy nhiên, có một điểm khác biệt quan trọng: nếu key không được tìm thấy, phương thức này trả về `None` thay vì phát sinh `KeyError`.

- `None` là một giá trị đặc biệt trong Python, đại diện cho việc không có giá trị.

```python
student_grades = {'Alice': 85, 'Bob': 92, 'Charlie': 78}

# Get Bob's grade (key exists)
bob_grade = student_grades.get('Bob')
print(f"Bob's grade (using get): {bob_grade}") # Output: Bob's grade (using get): 92

# Try to get David's grade (key does not exist)
david_grade = student_grades.get('David')
print(f"David's grade (using get): {david_grade}") # Output: David's grade (using get): None
```

- Bạn cũng có thể cung cấp một **giá trị mặc định (default value)** cho `get()` dưới dạng đối số thứ hai. Nếu key không được tìm thấy, `get()` sẽ trả về giá trị mặc định này thay vì `None`.

```python
student_grades = {'Alice': 85, 'Bob': 92, 'Charlie': 78}

# Get David's grade, return 0 if not found
david_grade = student_grades.get('David', 0)
print(f"David's grade (with default): {david_grade}") # Output: David's grade (with default): 0

# Get Alice's grade (key exists, default is ignored)
alice_grade = student_grades.get('Alice', 0)
print(f"Alice's grade (with default): {alice_grade}") # Output: Alice's grade (with default): 85
```

- Sử dụng `get()` thường được ưu tiên khi bạn không chắc một key có tồn tại hay không và muốn chương trình tiếp tục chạy mà không bị gián đoạn.

### **2.2. Checking for Key Existence with `in`**

- Một cách khác để tránh `KeyError` là kiểm tra xem một key có tồn tại trong Dictionary hay không trước khi truy cập nó.

- Bạn có thể thực hiện điều này bằng toán tử `in`. Toán tử này trả về `True` nếu key tồn tại và `False` nếu key không tồn tại.

```python
student_grades = {'Alice': 85, 'Bob': 92, 'Charlie': 78}

# Check if 'Alice' is a key
if 'Alice' in student_grades:
    print("Alice is in the dictionary.")
    # Safe to access now
    print(f"Alice's grade: {student_grades['Alice']}")
else:
    print("Alice is not in the dictionary.")

# Check if 'David' is a key
if 'David' in student_grades:
    print("David is in the dictionary.")
else:
    print("David is not in the dictionary.") # This will be printed
```

## **3. Adding or Modifying Key-Value Pairs**


- Việc thêm một cặp key-value mới hoặc thay đổi value được liên kết với một key đã tồn tại đều sử dụng cùng một cú pháp gán đơn giản với dấu ngoặc vuông `[]`.

- Nếu key được chỉ định trong dấu ngoặc vuông `[]` chưa tồn tại trong Dictionary, một cặp key-value mới sẽ được thêm vào. Nếu key đã tồn tại, value được liên kết với key đó sẽ được cập nhật thành value mới.

```python
student_grades = {'Alice': 85, 'Bob': 92, 'Charlie': 78}
print(f"Original grades: {student_grades}")

# Add a new student, David
student_grades['David'] = 88
print(f"After adding David: {student_grades}")

# Update Bob's grade
student_grades['Bob'] = 95
print(f"After updating Bob's grade: {student_grades}")
```

- Kết quả này giúp Dictionary trở nên linh hoạt lưu trữ dữ liệu có thể thay đổi hoặc mở rộng theo thời gian.

## **4. Removing Key-Value Pairs**

- Đôi khi bạn cần xóa các phần tử khỏi Dictionary. Python cung cấp một số cách để thực hiện việc này.

### **4.1. Using the del Statement**

- Câu lệnh `del` xóa một cặp key-value dựa trên key của nó.

```python
student_grades = {'Alice': 85, 'Bob': 95, 'Charlie': 78, 'David': 88}
print(f"Grades before deletion: {student_grades}")

# Remove Charlie from the dictionary
del student_grades['Charlie']
print(f"Grades after deleting Charlie: {student_grades}")
```

- Tương tự như khi truy cập bằng `[]`, nếu bạn sử dụng `del` với một key không tồn tại, Python sẽ phát sinh `KeyError`.

### **4.2. Using the `pop()` Method**

- Phương thức `pop()` cũng xóa một cặp key-value dựa trên key, nhưng phương thức này thực hiện thêm một thao tác: nó trả về value được liên kết với key vừa bị xóa.

- Điều này có thể hữu ích nếu bạn muốn sử dụng ngay value vừa bị xóa.

```python
student_grades = {'Alice': 85, 'Bob': 95, 'David': 88}
print(f"Grades before pop: {student_grades}")

# Remove David and store his grade
davids_removed_grade = student_grades.pop('David')
print(f"Removed David's grade: {davids_removed_grade}") # Output: Removed David's grade: 88
print(f"Grades after popping David: {student_grades}")  # Output: Grades after popping David: {'Alice': 85, 'Bob': 95}
```

- Nếu bạn sử dụng `pop()` với một key không tồn tại, phương thức này sẽ phát sinh `KeyError`, tương tự như `del`.

- Tuy nhiên, `pop()` cho phép bạn cung cấp một giá trị mặc định dưới dạng đối số thứ hai. Nếu key không được tìm thấy, `pop()` sẽ trả về giá trị mặc định thay vì phát sinh lỗi.

```python
student_grades = {'Alice': 85, 'Bob': 95}

# Try to pop 'Charlie', provide a default value
result = student_grades.pop('Charlie', 'Not Found')
print(f"Result of popping Charlie: {result}") # Output: Result of popping Charlie: Not Found
print(f"Grades remain unchanged: {student_grades}") # Output: Grades remain unchanged: {'Alice': 85, 'Bob': 95}
```

### **4.3. Using `popitem()`**

- Dictionary còn có một phương thức khác là `popitem()`. Phương thức này xóa và trả về một cặp `(key, value)` từ Dictionary.

- Trước Python 3.7, phương thức này xóa một phần tử bất kỳ. Từ Python 3.7 trở đi, nó tuân theo thứ tự **vào sau, ra trước (Last-In, First-Out — LIFO)**, nghĩa là phần tử được thêm gần nhất sẽ bị xóa trước.

- Phương thức này ít được người mới bắt đầu sử dụng hơn so với `del` hoặc `pop(key)`.

```python
student_grades = {'Alice': 85, 'Bob': 95}
student_grades['Eve'] = 90 # Add Eve last

# Remove and return the last added item (in Python 3.7+)
last_item = student_grades.popitem()
print(f"Removed item: {last_item}") # Output: Removed item: ('Eve', 90)
print(f"Dictionary after popitem: {student_grades}") # Output: Dictionary after popitem: {'Alice': 85, 'Bob': 95}
```

  - Thêm các cặp key-value mới
