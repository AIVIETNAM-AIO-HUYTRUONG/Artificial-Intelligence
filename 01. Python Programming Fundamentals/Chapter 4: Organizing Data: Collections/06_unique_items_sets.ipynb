# **Unique Items: Sets**

- Python cung cấp nhiều **kiểu tập hợp (collection types)** như list, tuple và dictionary. **set** là một kiểu collection khác, được thiết kế chuyên biệt để lưu trữ các phần tử duy nhất trong trường hợp thứ tự của các phần tử không quan trọng.

- Có thể hình dung set giống như **tập hợp toán học (mathematical set)**: một collection gồm các đối tượng khác nhau. Nếu bạn cố gắng thêm một phần tử đã tồn tại trong set, sẽ không có thay đổi nào xảy ra; set vẫn giữ nguyên vì nó chỉ lưu một bản sao của mỗi phần tử duy nhất. Thuộc tính này khiến set đặc biệt hữu ích cho các tác vụ như loại bỏ phần tử trùng lặp khỏi những collection khác hoặc nhanh chóng kiểm tra xem một phần tử có thuộc một nhóm hay không.

- Một đặc điểm xác định khác của set là chúng **không có thứ tự (unordered)**. Không giống list hoặc tuple, set không duy trì các phần tử theo bất kỳ trình tự cụ thể nào. Điều này có nghĩa là bạn không thể truy cập phần tử bằng index, chẳng hạn như `my_list[0]`, vì không tồn tại phần tử "đầu tiên" hoặc "thứ hai" được xác định.

## **1. Creating Sets**

- Bạn có thể tạo một set bằng cách đặt các phần tử được phân tách bằng dấu phẩy bên trong dấu ngoặc nhọn `{}`.

```python
# Creating a set with unique elements
unique_numbers = {1, 2, 3, 4, 5}
print(unique_numbers)
# Output might be {1, 2, 3, 4, 5} or some other order

# Creating a set with duplicates - duplicates are automatically removed
colors = {'red', 'green', 'blue', 'red'}
print(colors)
# Output: {'blue', 'red', 'green'} (order may vary)
```

- Hãy chú ý rằng `red` chỉ xuất hiện một lần trong set `colors`, mặc dù ban đầu chúng ta đã thêm nó hai lần.
- Một chi tiết nhỏ nhưng quan trọng: để tạo ra một set rỗng, bạn phải sử dụng hàm `set()` mà không truyền đối số. Việc sử dụng `{}` thực tế sẽ tạo ra một dictionary rỗng, như đã trình bày trước đó.

```python
# Correct way to create an empty set
empty_s = set()
print(type(empty_s))
# Output: <class 'set'>
print(empty_s)
# Output: set()

# This creates an empty dictionary, NOT a set
empty_d = {}
print(type(empty_d))
# Output: <class 'dict'>
```

- Bạn cũng có thể tạo ra một set từ bất kỳ **đối tượng có thể lặp (iterable object)**, chẳng hạn như list, tuple hoặc string, bằng cách sử dụng hàm `set()`. Đây là một kỹ thuật phổ biến để loại bỏ các phần tử trùng lặp.

```python
# Create a set from a list with duplicates
numbers_list = [1, 2, 2, 3, 4, 4, 4, 5]
unique_numbers_set = set(numbers_list)
print(unique_numbers_set)
# Output: {1, 2, 3, 4, 5} (order may vary)

# Create a set from a string
letters_set = set('hello')
print(letters_set)
# Output: {'h', 'l', 'o', 'e'} (order may vary)
```

## **2. Modifying Sets**

- Giống như list, set là **mutable**, nghĩa là bạn có thể thay đổi nội dung của chúng sau khi được tạo.

### **2.1. Adding Elements**

- Để thêm một phần tử vào set, sử dụng phương thức `.add()`

```python
permissions = {'read', 'write'}
print(permissions)
# Output: {'write', 'read'} (order may vary)

# Add a new permission
permissions.add('execute')
print(permissions)
# Output: {'write', 'read', 'execute'} (order may vary)

# Try adding an existing element - no change
permissions.add('read')
print(permissions)
# Output: {'write', 'read', 'execute'} (order may vary)
```

### **2.2 Removing Elements**

- Có hai cách chính để xóa phần tử:
  - `.remove(element)`: Phương thức này xóa phần tử được chỉ định. Tuy nhiên, nếu không tìm thấy phần tử trong set, nó sẽ phát sinh `KeyError`.
  - `.discard(element)`: Phương thức này cũng xóa phần tử được chỉ định, nhưng không phát sinh lỗi nếu không tìm thấy phần tử. Trong trường hợp đó, nó đơn giản là không thực hiện thay đổi nào.

- Việc lựa chọn giữa `.remove(element)` và `.discard(element)` phụ thuộc vào việc bạn muốn chương trình dừng lại khi cố gắng xóa một phần tử không tồn tại bằng `.remove()` hoặc muốn chương trình tiếp tục mà không thông báo lỗi bằng `.discard()`

```python
data_points = {10, 20, 30, 40}

# Remove an existing element using remove()
data_points.remove(20)
print(data_points)
# Output: {10, 40, 30} (order may vary)
```

```python
# Try removing a non-existent element using remove() - This will cause an error
data_points.remove(99) # Raises KeyError: 99
```

```python
# Remove an existing element using discard()
data_points.discard(30)
print(data_points)
# Output: {10, 40} (order may vary)

# Try removing a non-existent element using discard() - No error
data_points.discard(99)
print(data_points)
# Output: {10, 40} (order may vary)
```

## **3. Checking for Membership**

- Một trong những thao tác hiệu quả nhất của set là kiểm tra phần tử có tồn tại hay không bằng cách dùng `in`. Thao tác này nhanh hơn đáng kể so với việc kiểm tra membership trong list, đặc biệt với các collection lớn.

```python
allowed_users = {'alice', 'bob', 'charlie'}

# Check if a user is allowed
print('bob' in allowed_users)
# Output: True

print('eve' in allowed_users)
# Output: False

# Check if a user is NOT allowed
print('david' not in allowed_users)
# Output: True
```

## **4. When to Use Sets**

- Set là lựa chọn phù hợp khi:
  - **Yêu cầu duy nhất (uniqueness in required)**: Bạn cần lưu trữ một collection trong đó các phần tử trùng lặp không được phép tồn tại hoặc cần được tự động loại bỏ.

  - **Thường xuyên kiểm tra membership**: Chương trình thường xuyên cần kiểm tra xem một phần tử có tồn tại trong collection hay không. Set cung cấp khả năng tra cứu rất nhanh.

  - **Order Doesn't Matter**: Trình tự hoặc vị trí của các phần tử không quan trọng đối với ứng dụng.

  - **Mathematical Set Operations Needed**: Bạn dự định thực hiện các thao tác như tìm phần tử chung bằng **phép giao (intersection)**, kết hợp các phần tử bằng **phép hợp (union)** hoặc tìm sự khác biệt giữa các collection. Những thao tác này sẽ được trình bày trong phần tiếp theo.

- Ví dụ, nếu bạn có một list chứa ID người dùng từ các bản ghi log và muốn tìm những người dùng duy nhất đã truy cập hệ thống:

```python
log_entries = ['user1', 'user2', 'user1', 'user3', 'user2', 'user1']

# Use a set to find unique users
unique_users_set = set(log_entries)
print(unique_users_set)
# Output: {'user2', 'user3', 'user1'} (order may vary)

# If you need a list of unique users (order might not be original)
unique_users_list = list(unique_users_set)
print(unique_users_list)
# Output: ['user2', 'user3', 'user1'] (order may vary)
```

## **5. Conclusion**

- Set cung cấp một phương pháp mạnh mẽ và hiệu quả để xử lý các collection trong đó tính duy nhất và khả năng kiểm tra membership nhanh là những yêu cầu chính. Trong phần tiếp theo, chúng ta sẽ tìm hiểu các phép toán tập hợp cụ thể có thể thực hiện với set.  .
