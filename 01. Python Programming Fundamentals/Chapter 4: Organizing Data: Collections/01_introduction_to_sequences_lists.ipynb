# **Introduction to Sequences: Lists**

- Lập trình thường đòi hỏi làm việc với các **tập hợp (collections)** phần tử.

- Hãy tưởng tượng bạn cần theo dõi tên học sinh trong một lớp, các món đồ cần mua ở siêu thị, hoặc điểm số cao trong một trò chơi. Mặc dù bạn có thể tạo một biến riêng cho từng phần tử như:
  - `item1 = "milk"`,
  - `item2 = "eggs"`
  
  =>  Cách này sẽ nhanh chóng trở nên khó quản lý khi số lượng phần tử tăng lên.

- Python cung cấp một số **cấu trúc dữ liệu (data structures)** dựng sẵn, thường được gọi là **collection** hoặc **container**, được thiết kế riêng để lưu trữ nhiều dữ liệu cùng lúc. Một trong những cấu trúc cơ bản và linh hoạt nhất là **list**.

- Hãy coi list như một **dãy có thứ tự (ordered sequence)** các phần tử, giống như một list mua sắm hoặc một list việc cần làm được đánh số.
  - Mỗi phần tử trong list có một vị trí cụ thể.
  - Quan trọng là, list trong Python là **mutable (có thể thay đổi)**, nghĩa là bạn có thể chỉnh sửa nội dung của chúng sau khi đã tạo — bạn có thể thêm phần tử mới, xóa phần tử hiện có, hoặc thay đổi giá trị của phần tử đã có trong list

## **1. Creating a List**

- Bạn tạo một list trong Python bằng cách đặt một string các phần tử, phân tách bằng dấu (`,`), và nằm trong dấu (`[]`)

```python
# Một list rỗng
empty_list = []
print(empty_list)

# Một list số nguyên
numbers = [1, 2, 3, 5, 8]
print(numbers)

# Một list chuỗi
fruits = ["apple", "banana", "cherry"]
print(fruits)
# Một list chứa nhiều kiểu dữ liệu
mixed_list = [10, "hello", 3.14, True]
print(mixed_list)
```

- Như bạn thấy, lists rất linh hoạt và có thể chứa các phần tử thuộc nhiều **kiểu dữ liệu (data types)** khác nhau, bao gồm int, string, boolean, và thậm chí cả lists khác.

## **2. Accessing Items: Indexing**

- Vì lists có thứ tự, bạn có thể truy cập từng phần tử riêng lẻ bằng vị trí của nó, gọi là **chỉ số (index)**. Python sử dụng **zero-based indexing (đánh số bắt đầu từ 0)**, nghĩa là phần tử đầu tiên ở `index 0`, phần tử thứ hai ở `index 1`, và cứ tiếp tục như vậy.

- Để truy cập một phần tử, bạn dùng tên list rồi đặt `index` trong dấu (`[]`).

```python
fruits = ["apple", "banana", "cherry", "date"]

# Truy cập phần tử đầu tiên (chỉ số 0)
first_fruit = fruits[0]
print(first_fruit) # Kết quả: apple

# Truy cập phần tử thứ ba (chỉ số 2)
third_fruit = fruits[2]
print(third_fruit) # Kết quả: cherry
```

- Việc cố truy cập một index không tồn tại (ví dụ `fruits[4]` trong list ở trên, vốn chỉ có các index `0, 1, 2, 3`) sẽ dẫn đến lỗi **IndexError**.

## **3. Negative Indexing**

- Python cũng hỗ trợ **negative indexing**, đây là một cách tiện lợi để truy cập phần tử tính từ cuối list. index `-1` trỏ đến phần tử cuối cùng, `-2` trỏ đến phần tử áp chót, và cứ tiếp tục như vậy.

```python
fruits = ["apple", "banana", "cherry", "date"]

# Truy cập phần tử cuối cùng
last_fruit = fruits[-1]
print(last_fruit)  # Kết quả: date

# Truy cập phần tử áp chót
second_last_fruit = fruits[-2]
print(second_last_fruit) # Kết quả: cherry
```

## **4. Finding the Length of a List**

- Bạn có thể dễ dàng biết một list có bao nhiêu phần tử bằng **hàm dựng sẵn (built-in function)** `len()`.

```python
numbers = [10, 20, 30, 40, 50]
fruits = ["apple", "banana"]
empty_list = []

print(len(numbers)) # Kết quả: 5
print(len(fruits)) # Kết quả: 2
print(len(empty_list)) # Kết quả: 0
```

- list là một công cụ nền tảng để tổ chức dữ liệu trong Python. Những đặc điểm chính của nó là:

  - **Ordered (có thứ tự)**: Các phần tử giữ nguyên vị trí theo thứ tự được thêm vào.
  - **Mutable (có thể thay đổi)**: Bạn có thể thay đổi list sau khi tạo (thêm, xóa, sửa phần tử).
  - **Indexed (có chỉ số)**: Các phần tử có thể được truy cập trực tiếp bằng index số học, bắt đầu từ `0`.
  - **Heterogeneous (không đồng nhất về kiểu)**: Chúng có thể chứa các phần tử thuộc các kiểu dữ liệu khác nhau.

- Ở phần tiếp theo, chúng ta sẽ khám phá sâu hơn tính chất mutable của list và học cách thêm, xóa, và thay đổi các phần tử bên trong chúng.
