# **#6: Operations on Sets**

- **Sets** lưu các phần tử **unique (duy nhất)** và **unordered (không có thứ tự)**. Hãy xem một số phép toán mạnh mẽ bạn có thể thực hiện với chúng.

- Những phép toán này tương tự như **set theory (lý thuyết tập hợp)** trong toán học và rất hữu ích khi muốn so sánh hoặc kết hợp các collections dựa trên nội dung của chúng.

- Python cung cấp các **operators (toán tử)** trực quan và các methods tương ứng để thực hiện những thao tác này.

- Chúng ta sẽ dùng hai **set** ví dụ sau:

```python
programmers = {'Alice', 'Bob', 'Charlie', 'David'}
data_scientists = {'Charlie', 'David', 'Eve', 'Frank'}

print(f"Programmers: {programmers}")
print(f"Data Scientists: {data_scientists}")
```

## **1. Union Combining Sets**

- **Union (phép hợp)** của hai set tạo ra một set mới chứa tất cả các phần tử unique từ cả hai set gốc.

- Bạn có thể tính **union** bằng toán tử dấu `|` hoặc method `union()`.



```python
# Using the | operator
all_staff = programmers | data_scientists
print(f"Union (|): {all_staff}")

# Using the union() method
all_staff_method = programmers.union(data_scientists)
print(f"Union (method): {all_staff_method}")

# Note: The order doesn't matter for union
print(f"Is programmers | data_scientists == data_scientists | programmers? {programmers | data_scientists == data_scientists | programmers}")
```

- Cả hai cách đều cho cùng một kết quả: `{'Frank', 'Alice', 'Eve', 'Charlie', 'Bob', 'David'}`. Set này chứa tất cả những người là `programmer`, data `scientist`, hoặc cả hai.

## **2. Intersection: Finding Common Elements**

- **Intersection (phép giao)** của hai set tạo ra một set mới chỉ chứa các phần tử xuất hiện trong cả hai set gốc.

- Bạn có thể tính **intersection** bằng toán tử dấu `&` hoặc method `intersection()`.

```python
# Using the & operator
common_roles = programmers & data_scientists
print(f"Intersection (&): {common_roles}")

# Using the intersection() method
common_roles_method = programmers.intersection(data_scientists)
print(f"Intersection (method): {common_roles_method}")
```

- Kết quả trong cả hai trường hợp là `{'Charlie', 'David'}`, vì đây là những người xuất hiện trong cả set `programmers` và `data_scientists`.

## **3. Difference: Finding Unique Elements in One Set**

- **Difference (phép hiệu)** giữa hai set, giả sử **set A** và **set B**, tạo ra một set mới chứa các phần tử có trong A nhưng không có trong B. Điều quan trọng là thứ tự có ý nghĩa ở đây: A - B khác với B - A.

- Bạn có thể tính **difference** bằng toán tử trừ `-` hoặc method `difference()`.

```python
# Elements in programmers but NOT in data_scientists
only_programmers = programmers - data_scientists
print(f"Difference (programmers - data_scientists): {only_programmers}")

# Using the difference() method
only_programmers_method = programmers.difference(data_scientists)
print(f"Difference (method): {only_programmers_method}")

# Elements in data_scientists but NOT in programmers
only_data_scientists = data_scientists - programmers
print(f"Difference (data_scientists - programmers): {only_data_scientists}")
```

- Kết quả là:

  - `programmers` - `data_scientists`: `{'Alice', 'Bob'}`: Những programmer không đồng thời là data scientist
  - `data_scientists` - `programmers`: `{'Eve', 'Frank'}`: Nnhững data scientist không đồng thời là programmer

## **4. Symmetric Difference: Finding Elements in Exactly One Set**

- **Symmetric difference (phép hiệu đối xứng)** của hai set tạo ra một set mới chứa các phần tử có trong một trong hai set, nhưng không có trong cả hai.

- Về cơ bản, nó là phần ngược lại của **intersection**; nó bao gồm các phần tử duy nhất của từng set.

- Bạn có thể tính **symmetric difference** bằng toán tử dấu mũ `^` hoặc method `symmetric_difference()`

```python
# Using the ^ operator
exclusive_roles = programmers ^ data_scientists
print(f"Symmetric Difference (^): {exclusive_roles}")

# Using the symmetric_difference() method
exclusive_roles_method = programmers.symmetric_difference(data_scientists)
print(f"Symmetric Difference (method): {exclusive_roles_method}")
```

- Cả hai method đều cho kết quả `{'Frank', 'Alice', 'Eve', 'Bob'}`.

- Set này chứa những người chỉ là programmer hoặc chỉ là data scientist, không bao gồm những người làm cả hai vai trò.

## **5. Conclusion**

- **Sets** là cấu trúc dữ liệu mạnh mẽ cho phép thực hiện các phép toán logic và tập hợp tương tự như trong toán học. Các phép toán này rất hữu ích để xử lý, kết hợp, so sánh và lọc các tập hợp dữ liệu dựa trên nội dung của chúng.

- **1. Union (Phép hợp)**:
  - **Mục đích**: Kết hợp tất cả các phần tử duy nhất từ hai hoặc nhiều set thành một set mới. Nó trả về một set chứa mọi phần tử có mặt trong bất kỳ set nào đã cho.
  - **Toán tử**: `|`
  - **Method**: `.union()`
  - **Ví dụ khái niệm**: Nếu `Set A = {1, 2, 3}` và `Set B = {3, 4, 5}`, thì `A | B` (hoặc `A.union(B)`) sẽ là `{1, 2, 3, 4, 5}`. Phần tử trùng lặp (số 3) chỉ xuất hiện một lần.

- **2. Intersection (Phép giao)**:
  - **Mục đích**: Tìm kiếm các phần tử chung, tức là chỉ trả về những phần tử xuất hiện trong **tất cả** các set được so sánh.
  - **Toán tử**: `&`
  - **Method**: `.intersection()`
  - **Ví dụ khái niệm**: Nếu `Set A = {1, 2, 3}` và `Set B = {3, 4, 5}`, thì `A & B` (hoặc `A.intersection(B)`) sẽ là `{3}`. Đây là phần tử duy nhất có mặt trong cả hai set.

- **3. Difference (Phép hiệu)**:
  - **Mục đích**: Tìm các phần tử chỉ có trong set thứ nhất nhưng không có trong set thứ hai. Nó giúp xác định những yếu tố độc quyền của một set so với set khác.
  - **Toán tử**: `-`
  - **Method**: `.difference()`
  - **Lưu ý**: Thứ tự của các set trong phép hiệu là rất quan trọng. `A - B` không giống với `B - A`.
  - **Ví dụ khái niệm**: Nếu `Set A = {1, 2, 3}` và `Set B = {3, 4, 5}`,
    - `A - B` (hoặc `A.difference(B)`) sẽ là `{1, 2}` (các phần tử trong A không có trong B).
    - `B - A` (hoặc `B.difference(A)`) sẽ là `{4, 5}` (các phần tử trong B không có trong A).

- **4. Symmetric Difference (Phép hiệu đối xứng)**:
  - **Mục đích**: Trả về một set mới chứa tất cả các phần tử có mặt trong một trong hai set, nhưng **không phải trong cả hai**. Nó là tập hợp các phần tử độc quyền của mỗi set, loại bỏ đi những phần tử chung.
  - **Toán tử**: `^`
  - **Method**: `.symmetric_difference()`
  - **Ví dụ khái niệm**: Nếu `Set A = {1, 2, 3}` và `Set B = {3, 4, 5}`, thì `A ^ B` (hoặc `A.symmetric_difference(B)`) sẽ là `{1, 2, 4, 5}`. Đây là tổng hợp các phần tử độc nhất của A và B, không bao gồm phần tử chung là 3.
