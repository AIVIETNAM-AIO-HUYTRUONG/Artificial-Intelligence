# **Modifying Lists: Adding, Removing, Changing Items**

- Một trong những đặc điểm định nghĩa của list trong Python là **tính có thể thay đổi (mutability)**. Không giống như string hoặc **bộ dữ liệu bất biến (tuple)**, các phần tử trong list có thể được thay đổi, thêm vào, hoặc xóa đi sau khi list đã được tạo. Tính linh hoạt này khiến list trở thành công cụ cực kỳ hữu ích để lưu trữ các tập hợp dữ liệu có thể phát triển trong quá trình chương trình đang chạy.


## **1. Changing an Item at a Specific Index**

- Nếu bạn biết index của phần tử cần thay đổi, bạn có thể gán trực tiếp một giá trị mới vào index đó. Hãy nhớ rằng chỉ số của danh sách bắt đầu từ `0`.

```python
# Một danh sách màu sắc
colors = ["red", "green", "blue"]
print(f"Original list: {colors}")

# Đổi phần tử tại chỉ số 1 (phần tử thứ hai) từ "green" thành "yellow"
colors[1] = "yellow"
print(f"After changing index 1: {colors}")

# Đổi phần tử cuối cùng bằng chỉ số âm
colors[-1] = "purple"
print(f"After changing index -1: {colors}")
```

- Phép gán trực tiếp này sẽ thay thế giá trị hiện có tại index được chỉ định bằng giá trị mới. Nếu bạn dùng một index không tồn tại trong danh sách, Python sẽ tạo ra lỗi **IndexError**.

## **2. Adding Items to a List**

- Python cung cấp nhiều **phương thức (methods)** để thêm phần tử mới vào list.

### **2.1. Adding One Item to the End: `append()`**

- Cách phổ biến nhất để thêm một phần tử đơn lẻ là dùng phương thức `append()`.
- Phương thức này thêm phần tử vào cuối danh sách.

```python
# Một danh sách công việc trong ngày
tasks = ["email team", "review report"]
print(f"Tasks to do: {tasks}")

# Thêm một công việc mới vào cuối danh sách
tasks.append("buy groceries")
print(f"After append: {tasks}")
```

- `append()` rất đơn giản và hiệu quả khi bạn muốn mở rộng danh sách từng phần tử một.

### **2.2. Adding One Item at a Specific Position: `insert()`**

- Nếu bạn cần chèn một phần tử vào một vị trí cụ thể thay vì ở cuối, hãy dùng phương thức `insert()`.
- Method này nhận hai đối số:

  - index nơi bạn muốn chèn item
  - Là item muốn thêm

```python
# Chèn một công việc ưu tiên cao vào đầu danh sách (chỉ số 0)
tasks.insert(0, "call client")
print(f"After insert at index 0: {tasks}")

# Chèn thêm một công việc trước 'buy groceries' (hiện ở chỉ số 3)
tasks.insert(3, "prepare meeting notes")
print(f"After insert at index 3: {tasks}")
```

- Việc dùng `insert()` có thể kém hiệu quả hơn `append()` đối với list lớn, vì việc dời các phần tử cần thời gian.

### **2.3. Adding Multiple Items from Another Sequence: `extend()`**

- Để thêm toàn bộ items từ một **iterable (đối tượng có thể lặp)** khác, như list, tuple, hoặc string, vào cuối list hiện tại, hãy dùng `extend()`.

```python
# Thêm nhiều công việc từ một danh sách khác
more_tasks = ["schedule follow-up", "submit expenses"]
tasks.extend(more_tasks)
print(f"After extend: {tasks}")
```

#### **2.3.1. `extend()` and `append()`**

- Nếu bạn dùng `append()` với một list, toàn bộ list đó sẽ được thêm vào như một phần tử duy nhất. Còn `extend()` sẽ thêm từng phần tử riêng lẻ của **iterable** đó.

```python
numbers = [1, 2, 3]
extra_numbers = [4, 5]

# Dùng extend (cách đúng để thêm từng phần tử riêng lẻ)
numbers.extend(extra_numbers)
print(f"After extend: {numbers}") # Kết quả: [1, 2, 3, 4, 5]

# Đặt lại numbers và thử append
numbers = [1, 2, 3]
numbers.append(extra_numbers)
print(f"After append: {numbers}") # Kết quả: [1, 2, 3, [4, 5]] - Lưu ý đây là một danh sách lồng nhau!
```

## **3. Removing Items from a List**

- Cũng như có nhiều cách để thêm phần tử, có nhiều cách để xóa phần tử.

### **3.1. Removing the First Occurrence of a Value: `remove()`**

- Nếu bạn biết value của item muốn xóa nhưng không nhất thiết biết index, hãy dùng `remove()`.
Method này tìm trong list và xóa lần xuất hiện đầu tiên của value đó.
Nếu value bạn muốn xóa không có trong list, Python sẽ báo lỗi **ValueError**.

```python
# Quay lại với danh sách màu sắc của chúng ta
colors = ["red", "yellow", "purple", "yellow"]
print(f"Current colors: {colors}")

# Xóa lần xuất hiện đầu tiên của "yellow"
colors.remove("yellow")
print(f"After remove('yellow'): {colors}")
```

### **3.2. Removing an Item by Index: `pop()`**
- Nếu bạn muốn xóa item dựa trên index, hãy dùng `pop()`.
- Method này còn trả về item đã bị xóa, điều này rất hữu ích.
- Nếu bạn không chỉ định index, `pop()` sẽ xóa và trả về item cuối cùng trong list.
- Dùng `pop()` với một index không hợp lệ sẽ gây ra **IndexError**.

```python
print(f"Current colors: {colors}") # Dự kiến: ['red', 'purple', 'yellow']

# Xóa và lấy phần tử tại chỉ số 1 ("purple")
removed_color = colors.pop(1)
print(f"Removed color: {removed_color}")
print(f"After pop(1): {colors}")

# Xóa và lấy phần tử cuối cùng (hành vi mặc định)
last_color = colors.pop()
print(f"Removed last color: {last_color}")
print(f"After pop(): {colors}")
```

### **3.3. Removing an Item by Index or Slice: `del` statement**

- `del` statement là cách tổng quát hơn để xóa item hoặc một slice của list bằng index.

- Khác với `pop()`, `del` không trả về giá trị đã bị xóa.

```python
numbers = [10, 20, 30, 40, 50, 60]
print(f"Original numbers: {numbers}")

# Xóa phần tử tại chỉ số 0
del numbers[0]
print(f"After del numbers[0]: {numbers}")

# Xóa các phần tử từ chỉ số 2 đến trước chỉ số 4
# Điều này xóa các phần tử hiện đang ở chỉ số 2 và 3 (ban đầu là 30 và 40, giờ là 40 và 50)
del numbers[2:4]
print(f"After del numbers[2:4]: {numbers}")
```

```python
# del cũng có thể xóa toàn bộ biến danh sách, nhưng điều đó khác với việc làm rỗng danh sách
del numbers
print(numbers) # Khi đó sẽ gây ra lỗi NameError
```

### **3.4. Removing All Items: `clear()`**

- Nếu bạn muốn xóa toàn bộ items khỏi list và biến nó thành list rỗng, hãy dùng `clear()`.

```python
numbers = [10, 60] # Từ ví dụ trước
print(f"Numbers before clear: {numbers}")

numbers.clear()
print(f"Numbers after clear: {numbers}")
```

- Phương thức này sửa đổi danh sách **trực tiếp (in place)**, làm cho nó rỗng. Biến danh sách vẫn tồn tại, chỉ là nó không còn chứa phần tử nào.

  - **Thao tác "trực tiếp" (in-place)** trong Python là các thao tác sửa đổi đối tượng danh sách (list) hiện tại mà không tạo ra một bản sao mới trong bộ nhớ.

  - Khi bạn thực hiện một **phương thức in-place**, địa chỉ bộ nhớ (`id()`) của danh sách không thay đổi. Các thao tác này thường trả về None, vì đối tượng gốc đã bị thay đổi trực tiếp.

  - Các đặc điểm chính:
    - **Hiệu quả bộ nhớ**: Không cấp phát thêm vùng nhớ cho một danh sách mới.
    - **Tham chiếu**: Mọi biến khác đang tham chiếu đến cùng danh sách đó sẽ thấy sự thay đổi.
    - Ví dụ về các **phương thức in-place** phổ biến:
      - `list.append(x)`: Thêm phần tử vào cuối.
      - `list.extend(iterable)`: Mở rộng danh sách bằng cách thêm các phần tử từ iterable.
      - `list.insert(i, x)`: Chèn phần tử tại vị trí chỉ định.
      - `list.remove(x)`: Loại bỏ phần tử đầu tiên có giá trị bằng x.
      - `list.pop(i)`: Loại bỏ và trả về phần tử tại vị trí i.
      - `list.sort()`: Sắp xếp danh sách tại chỗ (khác với hàm `sorted()` tạo ra list mới).
      - `list.reverse()`: Đảo ngược thứ tự các phần tử tại chỗ.
