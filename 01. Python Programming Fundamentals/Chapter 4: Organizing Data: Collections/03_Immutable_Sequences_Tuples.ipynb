# **Immutable Sequences: Tuples**

- Danh sách là một **vùng chứa linh hoạt (flexible container)** dành cho dữ liệu có thứ tự. Các phần tử trong danh sách có thể được thêm, xóa hoặc thay đổi trong quá trình chương trình hoạt động.

- `Tuple` là một **kiểu chuỗi tích hợp sẵn (built-in sequence type)** khác với Python. Khi nhìn lần đầu, **tuple** có nhiều điểm tương đồng với **list**. Tuy nhiên, giữa hai cấu trúc tồn tài một khác biết quan trọng: tuple là **bất biến (immutable)**

- Điều này có nghĩa là sau khi một tuple được tạo, các phần tử bên trong nó không thể:
  - Bị thay đổi
  - Được bổ sung
  - Bị loại bỏ

- Có thể hình dung tuple như một **chuỗi phần tử cố định (fixed sequence of items)

- Nếu một tập hợp giá trị đại diện cho dữ liệu mang tính cố định, chẳng hạn như tọa độ của một điểm (x, y) hoặc các giá trị RGB của một màu cụ thể, tuple thường là lựa chọn phù hợp hơn danh sách.


## **1. Creating Tuples**

- Tuple thường được tạo bằng cách sử dụng **dấu ngoặc tròn (parentheses)** `()` thay cho dấu **ngoặc vuông (square brackets)** `[]` được sử dụng với danh sách.

- Các phần tử trong tuple được phân tách bằng dấu phẩy.

```python
# Một tuple rỗng
empty_tuple = ()
print(empty_tuple)

# Một tuple chứa các số nguyên
coordinates = (10, 20)
print(coordinates)

# Một tuple chứa nhiều kiểu dữ liệu nếu ngữ cảnh đã rõ ràng
person_info = ("Alice", 30, "New York")
print(person_info)

# Dấu ngoặc tròn đôi khi có thể được lược bỏ
another_tuple = 100, 200, 300
print(another_tuple)
print(type(another_tuple)) # Output: <class 'tuple'>
```

- Python vẫn tạo một tuple mặc dù các giá trị không được đặt bên trong dấu ngoặc tròn.

- Điều này cho thấy dấu phẩy mới là thành phần quan trọng xác định một tuple. Dấu ngoặc tròn chủ yếu được sử dụng để giúp cấu trúc mã nguồn rõ ràng hơn.

### **1.1. Creating a Single-Item Tuple**

- Việc tạo một tuple chỉ chứa một phần tử yêu cầu một cú pháp đặc biệt: cần đặt một **dấu phẩy ở cuối (trailing comma)** phần tử.

- Nếu không có dấu phẩy, Python sẽ hiểu dấu ngoặc tròn là dấu ngoặc dùng để nhóm biểu thức, thay vì dùng để định nghĩa tuple.



```python
# Đây KHÔNG phải là tuple, Nó chỉ là số nguyên 5 được đặt trong dấu ngoặc
not_a_tuple = (5)
print(type(not_a_tuple)) # Output: <class 'int'>

# Đây là một tuple chứa một phần tử
single_item_tuple = (5,)
print(type(single_item_tuple)) # Output: <class 'tuple'>
print(single_item_tuple)
```

- **Dấu phẩy `,`** phía sau số `5` nói cho Python biết rõ rằng lập trình viên muốn tạo một tuple, ngay cả khi cấu trúc này chỉ chứa một phần tử.

## **2. Accessing Tuple Elements**

- Cách truy cập phần tử của tuple hoàn toàn tương tự danh sách. Chúng ta sử dụng dấu ngoặc vuông `[]` cùng với index của phần tử cần truy xuất. Cần nhớ rằng chỉ mục trong Python bắt đầu từ `0`.

```python
rgb_color = (255, 165, 0) # Orange color

# Truy cập phần tử đầu tiên tại chỉ mục 0
red_value = rgb_color[0]
print(f"Red: {red_value}") # Output: Red: 255

# Truy cập phần tử thứ ba tại chỉ mục 2
blue_value = rgb_color[2]
print(f"Blue: {blue_value}") # Output: Blue: 0
```

### **2.1. Slicing in Tuple**

- Tuple cũng hỗ trợ **phép cắt (slicing)** tương tự danh sách. Phép cắt cho phép lấy một phần của tuple để tạo thành một chuỗi con.

- Cú pháp cơ bản là: `tuple[start:stop]`
  - `start` là chỉ mục bắt đầu và được bao gồm trong kết quả.
  - `stop` là chỉ mục kết thúc nhưng không được bao gồm trong kết quả.

```python
weekdays = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")

# Lấy các phần tử từ index 1 đến trước index 4
mid_week = weekdays[1:4]
print(mid_week) # Output: ('Tue', 'Wed', 'Thu')

# Lấy hai phần tử cuối cùng
weekend = weekdays[-2:]
print(weekend) # Output: ('Sat', 'Sun')
```

## **3. Immutability: The Defining Feature**

- Tính bất biến (immutability) là đặc điểm quan trọng nhất phân biệt tuple với danh sách.

- Sau khi một tuple được tạo, chúng ta không thể thay đổi trực tiếp các phần tử bên trong nó. Nếu cố gắng thay đổi giá trị thì Python sẽ báo lỗi **TypeError**
- Tương tự, tuple không cung cấp các phương thức thay đổi kích thước hoặc nội dung như: `append()`, `extend()`, `remove()`, `pop()`. Các phương thức trên tồn tại đối với danh sách nhưng không tồn tại đối với tuple.

```python
my_tuple = (10, 20, 30)
print(my_tuple)

# Let's try to change the element at index 1
# This will cause an error!
my_tuple[1] = 25 # Uncommenting this line leads to TypeError

# Similarly, you cannot add or remove elements
# Methods like append(), extend(), remove(), pop() do not exist for tuples
```

### **3.1. Why is immutability useful**

- **Bảo toàn tính toàn vẹn của dữ liệu (Data Integrity)**:
  - Khi sử dụng tuple để biểu diễn dữ liệu không được phép thay đổi, chúng ta có thể đảm bảo rằng dữ liệu đó giữ nguyên trong suốt quá trình chương trình hoạt động. Điều này giúp ngăn chặn các thay đổi ngoài ý muốn.
  - Tuple thường phù hợp để biểu diễn:
    - Tọa độ cố định
    - Các giá trị màu RGB
    - Thông số cấu hình không thay đổi
    - Kích thước của một đối tượng
    - Một bản ghi dữ liệu cố định
  - VD: `screen_resolution = (1920, 1080)`. Tuple trên đại diện cho độ phân giải màn hình gồm chiều rộng và chiều cao. Nếu các giá trị này cần được giữ cố định, tuple giúp thể hiện rõ mục đích đó trong mã nguồn.

- **Sử dụng làm khóa của từ điển (Dictionary Keys)**:
  - **Các khóa của từ điển (dictionary keys)** phải là những đối tượng có thể băm được, thường yêu cầu chúng phải **immutable**.
  - Tuple có thể được sử dụng làm khóa của từ điển, với điều kiện tất cả phần tử bên trong tuple cũng là các đối tượng bất biến, chẳng hạn như:

    - Số
    - Chuỗi
    - Tuple bất biến khác

  - Danh sách không thể được sử dụng làm khóa vì danh sách có thể thay đổi.


```python
# Sử dụng tuple làm khóa của từ điển
location_temps = {
    (40.7128, -74.0060): 22,  # New York City coordinates -> temperature
    (34.0522, -118.2437): 28, # Los Angeles coordinates -> temperature
}

print(location_temps[(40.7128, -74.0060)]) # Output: 22
```

```python
# Trying to use a list as a key would cause a TypeError
location_temps[[34.0522, -118.2437]] = 28 # Error!
```

- **Performance**:
  - Tính bất biến (immutability) và kích thước cố định của `tuple` cho phép trình thông dịch Python thực hiện các tối ưu hóa cấp thấp mà `list` không thể làm được:
  
    1. **Cấp phát bộ nhớ (Memory Allocation)**:
        - Khi tạo một `tuple`, Python biết chính xác kích thước cần thiết và cấp phát một khối bộ nhớ liền kề duy nhất.
        -  Ngược lại, `list` là các mảng động (dynamic arrays) yêu cầu thêm bộ nhớ dự phòng để phục vụ các thao tác append hoặc extend, dẫn đến lãng phí bộ nhớ và **chi phí tái cấp phát (reallocation)** khi mảng đầy.
    2. **Cơ chế lưu trữ tạm thời (Free List/Caching)**:
        - Python duy trì một danh sách các `tuple` nhỏ đã được giải phóng để tái sử dụng ngay lập tức cho các yêu cầu mới, giúp giảm áp lực lên **bộ quản lý bộ nhớ (memory allocator)**.
    3. **Tra cứu và truy cập**:
        - Do cấu trúc tĩnh, việc truy cập các phần tử trong `tuple` có độ trễ thấp hơn. Ngoài ra, vì `tuple` là hashable (nếu các phần tử bên trong đều hashable), nó có thể được sử dụng làm khóa (key) trong dict hoặc phần tử trong set, điều mà list không thể thực hiện do tính chất có thể thay đổi.

## **4. Tuple Methods**

- Do tuple là bất biến, nó có ít **phương thức tích hợp sẵn (built-in methods)** hơn đáng kể so với danh sách.

- Hai phương thức chính của tuple là:
  - `count(value)`: trả về số lần một giá trị xuất hiện trong tuple
  - `index(value)`: trả về chỉ mục của lần xuất hiện đầu tiên của một giá trị trong tuple.

```python
numbers = (1, 2, 5, 2, 8, 2, 10)

# Count occurrences of the number 2
count_of_2 = numbers.count(2)
print(f"The number 2 appears {count_of_2} times.") # Output: The number 2 appears 3 times.

# Find the index of the first occurrence of 8
index_of_8 = numbers.index(8)
print(f"The number 8 first appears at index {index_of_8}.") # Output: The number 8 first appears at index 4.

# Trying to find a value not in the tuple raises an error
# index_of_99 = numbers.index(99) # Uncommenting this line leads to ValueError
```

## **5. Tuple Packing and Unpacking**

- Python cung cấp một cú pháp thuận tiện liên quan đến tuple, được gọi là:

### **5.1. Packing**

-  Khi nhiều giá trị được viết cách nhau bằng dấu phẩy, Python có thể tự động đóng gói chúng thành một tuple.

```python
packed_data = "Max", 25, "Berlin" # Values are packed into a tuple
print(packed_data)      # Output: ('Max', 25, 'Berlin')
print(type(packed_data)) # Output: <class 'tuple'>
```

### **5.2. Unpacking**

- **Giải nén (unpacking**) cho phép gán các phần tử của một tuple cho nhiều biến riêng biệt trong một câu lệnh duy nhất

- Theo cách hiểu đơn giản, các phần tử trong tuple ở phía bên phải được phân phối lần lượt cho các biến ở phía bên trái.

- Thông thường, số lượng biến ở bên trái phải bằng số lượng phần tử trong tuple

```python
point = (15, 40)

# Unpack the tuple into variables x and y
x, y = point

print(f"x coordinate: {x}") # Output: x coordinate: 15
print(f"y coordinate: {y}") # Output: y coordinate: 40

# This is useful for swapping variables too!
a = 10
b = 20
a, b = b, a # Pack (b, a) into a tuple, then unpack into a, b
print(f"a: {a}, b: {b}") # Output: a: 20, b: 10
```

```python
x, y, z = point
print(f"x: {x}, y: {y}, z: {z}")
```

- **extended unpacking**: Python cũng hỗ trợ giải nén mở rộng (extended unpacking) bằng toán tử `*`.

```python
numbers = (10, 20, 30, 40, 50)

first, *middle, last = numbers
print(first)
print(middle)
print(last)
```

- Cần lưu ý rằng biến sử dụng toán tử `*` nhận kết quả dưới dạng danh sách, không phải tuple.

## **6. Choosing Between Lists and Tuples**

- Danh sách và tuple đều là các kiểu chuỗi (sequence types), có khả năng:
  - Lưu trữ nhiều phần tử
  - Duy trì thứ tự của các phần tử
  - Hỗ trợ chỉ mục
  - Hỗ trợ cắt chuỗi
  - Cho phép lặp qua từng phần tử

- Tuy nhiên, quyết định lựa chọn danh sách hay tuple phụ thuộc vào việc dữ liệu có cần thay đổi sau khi được tạo hay không.

- **Khi nào nên sử dụng List?**
  - Nên sử dụng danh sách khi tập hợp phần tử có thể cần được thay đổi trong quá trình chương trình hoạt động.

  - Các thao tác có thể bao gồm:

    - Thêm phần tử
    - Xóa phần tử
    - Cập nhật giá trị
    - Sắp xếp lại dữ liệu
    - Mở rộng hoặc thu nhỏ tập hợp

- **Khi nào nên sử dụng Tuple?**
  - Nên sử dụng tuple khi tập hợp phần tử đại diện cho một thực thể cố định và không nên thay đổi sau khi được tạo.

  - Ví dụ phù hợp với tuple:

    - Tọa độ của một điểm
    - Giá trị màu RGB
    - Ngày, tháng và năm
    - Kích thước chiều rộng và chiều cao
    - Các thiết lập cấu hình cố định
    - Một bản ghi không được phép thay đổi
