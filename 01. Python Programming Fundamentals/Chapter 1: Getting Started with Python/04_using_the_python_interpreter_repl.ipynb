# **Using the Python Interpreter (REPL)**

- Trình thông dịch tương tác của Python, thường được gọi là **REPL**, cung cấp một cách trực tiếp để tương tác với Python. REPL là viết tắt của **Read – Eval – Print – Loop**, mô tả khá chính xác chức năng của nó: nó đọc mã Python mà bạn nhập vào, đánh giá và thực thi mã đó, in kết quả ra màn hình nếu có, rồi quay lại trạng thái chờ để nhận đầu vào tiếp theo.

- Đây là một môi trường rất hữu ích để thử nghiệm các đoạn mã nhỏ, kiểm tra cú pháp hoặc khám phá khả năng của Python mà không cần tạo một tệp tin riêng.

## **1. What is the REPL?**

- Có thể hình dung REPL như một cuộc đối thoại trực tiếp với trình thông dịch Python. Bạn nhập một lệnh, nhấn "Enter", và Python phản hồi ngay lập tức. Vòng phản hồi tức thì này đặc biệt hữu ích cho việc học tập và kiểm tra nhanh.

  - **Read**: Đọc dòng mã bạn nhập vào.
  - **Eval**: Đánh giá và thực thi đoạn mã đó.
  - **Print**: In kết quả của quá trình đánh giá ra màn hình.
  - **Loop**: Chờ đầu vào tiếp theo từ bạn.

## **2. Starting the REPL**

- Thông thường, bạn có thể khởi động REPL của Python bằng cách mở **terminal** hoặc **command prompt**, sau đó nhập `python` hoặc đôi khi là `python3`, tùy thuộc vào hệ điều hành và cách cài đặt, rồi nhấn "Enter".

  - **Windowsa**: Mở **Command Prompt** (tìm `cmd`) hoặc **PowerShell** -> Nhập `python` và nhấn Enter.

  - **macOS**: Mở **Terminal** (Applications > Utilities > Terminal, hoặc dùng Spotlight để tìm) -> Nhập `python3` (thường được ưu tiên trên macOS để phân biệt với Python 2 cũ của hệ thống) và nhấn Enter.

  - **Linux**: Mở ứng dụng **Terminal** -> Nhập `python3` (hoặc python nếu `python3` đã được liên kết với `python`) và nhấn Enter.

- Nếu mọi thứ được thiết lập đúng, bạn sẽ thấy giao diện tương tự như sau, dù số phiên bản và chi tiết cụ thể có thể khác:

    ```py
    Python 3.10.4 (main, Mar 31 2022, 08:41:55) [GCC 7.5.0] on linux
    Type "help", "copyright", "credits" or "license" for more information.
    >>>
    ```

- Dấu nhắc `>>>` là **Python prompt**. Nó cho biết trình thông dịch đã sẵn sàng nhận lệnh của bạn.

## **3. Why Use the REPL?**

- REPL đặc biệt hữu ích trong các trường hợp sau:

  - **Thử nghiệm nhanh**: Kiểm tra một đoạn mã nhỏ hoặc cú pháp mà không cần tạo tệp tin.
  - **Khám phá đối tượng**: Xem các phương thức hoặc thuộc tính có sẵn của một biến hay đối tượng bằng các hàm như `dir()` và `help()`.
  - **Debugging**: Chạy nhanh một hoặc hai dòng mã để quan sát hành vi của chương trình.
  - **Học tập**: Nhận phản hồi ngay lập tức về cú pháp và lệnh trong Python.

## **4. Using the REPL**

- Hãy thử một số lệnh đơn giản trực tiếp tại dấu nhắc `>>>.`

  - **Basic Arithmetic**: Bạn chỉ cần nhập một biểu thức toán học và nhấn Enter. Python sẽ đánh giá kết quả ngay lập tức.

    ```
    >>> 2 + 3
    5
    >>> 100 - 5 * 10
    50
    >>> (50 + 50) * 2 / 4
    50.0
    ```
  - **Assigning Variables**: Bạn có thể tạo biến trong REPL giống như khi viết trong một tệp script.

    ```
    >>> message = "Learning Python is fun"
    >>> print(message)
    Learning Python is fun
    >>> count = 10
    >>> count * 5
    50
    ```

    - Lưu ý rằng việc chỉ gán một biến như `message = ...` sẽ không tạo ra output. Tuy nhiên, nếu bạn chỉ nhập tên biến, REPL thường sẽ in giá trị của nó:
    ```
    >>> message
    'Learning Python is fun'
    >>> count
    10
    ```

## **5. Exiting the REPL**

- Khi bạn hoàn tất việc sử dụng trình thông dịch tương tác, có thể thoát bằng một trong các cách sau:

  - Gõ `exit()` rồi nhấn Enter.
  - Gõ `quit()` rồi nhấn Enter.
  - Nhấn `Ctrl+D` trên Linux/macOS hoặc `Ctrl+Z` rồi nhấn Enter trên Windows.

Sau đó, bạn sẽ quay trở lại terminal hoặc command prompt bình thường.

- Mặc dù phần lớn chương trình sẽ được viết trong các tệp script (`.py`).

- REPL vẫn là một công cụ hỗ trợ rất hữu ích xuyên suốt quá trình phát triển Python. Nó giống như một “môi trường thử nghiệm” để bạn kiểm tra ý tưởng nhanh chóng.
