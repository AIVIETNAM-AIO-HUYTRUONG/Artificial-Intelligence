# **What is Python Programming?**

- Lập trình (Programming) cho phép con người truyền đạt các chỉ dẫn đến máy tính. Về bản chất, máy tính chỉ hiểu các chuỗi số 0 và 1, thường được gọi là **mã máy (machine code)**. Tuy nhiên, việc viết trực tiếp bằng machine code là một quá trình cực kỳ phức tạp và tốn nhiều thời gian.

- Các **ngôn ngữ lập trình (programming languages)** đóng vai trò như một lớp trung gian, cung cấp phương thức biểu đạt gần gũi hơn với con người để mô tả các chỉ dẫn, sau đó được chuyển đổi sang dạng máy tính có thể thực thi

- Python là một trong những ngôn ngữ lập trình như vậy. Python được biết đến là một ngôn ngữ **bậc cao (high-level)**, **thông dịch (interpreted)** và **đa mục đích (general-purpose)**

  - **High-level**:
    
    - Điều này có nghĩa rằng cú pháp của Python được thiết kế gần với ngôn ngữ của con người hơn, đồng thời che giấu nhiều chi tiết kỹ thuật phức tạp liên quan đến phần cứng máy tính.
    - Có thể hình dung như việc đưa ra chỉ dẫn lái xe: Thay vì mô tả từng thao tác xoay vô lăng hay nhấn bàn đạp, bạn chỉ cần nói "Rẽ trái ở ngã tư tiếp theo"
    - Nhờ đặc điểm này, mã nguồn Python thường dễ đọc, dễ viết và dễ bảo trì hơn so với các ngôn ngữ có cấp thấp hơn như C hoặc Asembly

  - **Interpreted**:
    
    - Mã Python thường được lưu thực thi theo từng dòng thông qua một chương trình gọi là **trình thông dịch (interpreter)**
    - Khi chạy một tệp lệnh Python, trình thông dịch sẽ đọc mã nguồn và thực hiện các hành động tương ứng ngay lập tức. Điều này khác với các **ngôn ngữ biên dịch (compiled language)**, nơi toàn bộ chương trình phải được chuyển đổi thành mã máy trước khi có thể thực thi.
    - Đặc tính thông dịch thường giúp tăng tốc quá trình phát triển phần mềm, đặc biệt trong giai đoạn kiểm thử và gỡ lỗi, bởi lập trình viên không cần thực hiện bước biên dịch riêng biệt sau mỗi lần chỉnh sửa.

  - **General-Purpose**: Python không được thiết kế dành riêng cho một nhiệm vụ cụ thể. Đây là một ngôn ngữ linh hoạt và được ứng dụng trong nhiều lĩnh vực khác nhau:

    - Phát triển ứng dụng web (ví dụ: hệ thống phía máy chủ của website)
    - Phân tích dữ liệu
    - Trí tuệ nhân tạo (AI)
    - Tính toán khoa học
    - Tự động hóa tác vụ (scripting)
    - Và nhiều lĩnh vực khác

- Python được tạo ra vào cuối những năm 1980 bởi Guido van Rossum.

- Một trong những mục tiêu cốt lõi trong quá trình thiết kế Python là **tăng khả năng đọc hiểu mã nguồn (code readability)**. Vì vậy, cú pháp của Python được xây dựng theo hướng tối giản, rõ ràng và đặc biệt nhấn mạnh việc sử dụng **khoảng trắng (indentation)** để tổ chức các khối lệnh. Điều này góp phần đáng kể vào tính minh bạch và khả năng bảo trì của chương trình.

- Trong nhiều trường hợp, mã Python được viết tốt có thể được đọc gần giống như các câu văn tiếng Anh thông thường.

- Ví dụ, một câu lệnh đơn giản để hiển thị thông báo:

```python
print("Hello, Python learner!")
```

- Chỉ với một dòng lệnh, mục đích của chương trình đã được thể hiện rất rõ ràng. Trong các ngôn ngữ mức thấp hơn, để đạt được cùng kết quả có thể cần nhiều bước thiết lập hơn cùng cú pháp kém trực quan hơn.

## **Các Đặc Trưng Cốt lõi của Python - Core Characteristics of Python**

- Một số đặc điểm dưới đây góp phần tạo nên sự phổ biến của Python, đặc biệt đối với người mới bắt đầu:

  1. **Tính dễ đọc và đơn giản (Reqdability and Simplicity)**: Như đã đề cập, cú pháp rõ ràng giúp người học tập trung vào việc hiểu các nguyên lý lập trình thay vì bị phân tán bởi những quy tắc cú pháp phức tạp.

  2. **Thư viện chuẩn phong phú (Large Standard Library)**: Python đi kèm với một tập hợp lớn các modules được xây dựng sẵn (gọi là **thư viện chuẩn - standard library**) nhằm xử lý các tác vụ phổ biến như:
      - Làm việc với văn bản
      - Xử lý số liệu
      - Đọc và ghi tệp
      - Kết nối mạng
      - Và nhiều chức năng khác
  - Nhờ đó, lập trình viên thường không cần phải xây dựng mọi chức năng từ đầu.

  3. **Cộng đồng và hệ sinh thái mạnh mẽ (Vibrant Community and Ecosystem)**:

  - Python sở hữu một cộng đồng toàn cầu rộng lớn và năng động.

  - Điều này mang lại nhiều lợi ích như:

    - Tài nguyên học tập phong phú    
    - Diễn đàn hỗ trợ kỹ thuật    
    - Kho thư viện bên thứ ba khổng lồ có thể cài đặt thông qua các công cụ như `pip`   

  - Các thư viện này mở rộng khả năng của Python sang gần như mọi lĩnh vực ứng dụng có thể tưởng tượng.

  4. **Kiểu dữ liệu động (Dynamic Typing)**:

    - Trong Python, lập trình viên thường không cần khai báo trước kiểu dữ liệu của biến (ví dụ: số hay chuỗi ký tự).

    - Python sẽ tự động xác định kiểu dữ liệu tại thời điểm chương trình thực thi.

    - Mặc dù điều này giúp tăng tính linh hoạt, việc hiểu rõ các kiểu dữ liệu vẫn rất quan trọng và sẽ được nghiên cứu ở các bài tiếp theo.

- Sự kết hợp giữa tính đơn giản, sức mạnh và khả năng linh hoạt khiến Python trở thành lựa chọn lý tưởng cho ngôn ngữ lập trình đầu tiên. Đồng thời, nó cung cấp nền tảng kiến thức vững chắc có thể áp dụng trong nhiều lĩnh vực như phát triển phần mềm và khoa học dữ liệu.

- Khóa học này được thiết kế để đồng hành cùng người học ngay từ những bước khởi đầu, không giả định bất kỳ kinh nghiệm lập trình nào trước đó. Chúng ta sẽ tận dụng các ưu điểm của Python để tiếp cận và nắm vững các khái niệm lập trình một cách hiệu quả.
